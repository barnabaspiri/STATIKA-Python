##### Csomagok importálása

```python
import numpy as np # Numerikus számítások
import matplotlib.pyplot as plt # Ábrázolás
plt.rcParams['mathtext.fontset'] = 'cm' # Betűtípus az ábrákon
plt.rcParams['font.family'] = 'serif' # Betűtípus az ábrákon
plt.rcParams['font.serif'] = ['cmr10'] # Betűtípus az ábrákon
plt.rcParams['axes.formatter.use_mathtext'] = True # Betűtípus az ábrákon
plt.rcParams['axes.unicode_minus'] = False # Betűtípus az ábrákon
from IPython.display import display, Math, Latex # Szép kimenet Jupyter notebookban
import sympy as sp # Szimbólikus számítások
from sympy import latex # Szép kimenet Jupyter notebookban
sp.init_printing() # Szép kimenet Jupyter notebookban
```

##### Adatok

```python
a = 1 # m

F_1 = 400 # N
F_2 = 200 # N
```

Minden rudat (számozással) és minden csomópontot (betűkkel) azonosítsunk!

##### Szabadtest ábra

<img src="img/02.jpg" alt="Feladat" width=450px>

##### A reakciók

```python
A_x, A_y, B_y = sp.symbols('A_x, A_y, B_y')
```

##### Egyensúlyi egyenletek

```python
# x irányú erő egyensúly
eq1 = A_x - F_2
display(Math(rf'\sum F_x = 0 \quad \longrightarrow \quad {latex(eq1)} = 0'))

# y irányú erő egyensúly
eq2 = A_y - F_1 + B_y
display(Math(rf'\sum F_y = 0 \quad \longrightarrow \quad {latex(eq2)} = 0'))

# z irányú nyomatéki egyensúly
eq3 = -F_1*a + F_2*a + B_y*4*a
display(Math(rf'\sum M_A = 0 \quad \longrightarrow \quad {latex(eq3)} = 0'))
```

##### Egyensúlyi egyenletek megoldása

```python
sol = sp.solve([eq1, eq2, eq3], [A_x, A_y, B_y])

A_xsol = sol[A_x]
display(Math(rf'A_x = {latex(A_xsol)} ~ \mathrm{{N}}'))

A_ysol = sol[A_y]
display(Math(rf'A_y = {latex(A_ysol)} ~ \mathrm{{N}}'))

B_ysol = sol[B_y]
display(Math(rf'B_y = {latex(B_ysol)} ~ \mathrm{{N}}'))
```

##### A rúderők

* Mindig olyan csuklót válasszunk, ahol max. 2 ismeretlen van!

* Az ismeretlen rúderőket úgy vegyük fel, mintha a rúd húzott lenne! Így a pozitív rúderő húzott, a negatív rúderő nyomott rudat jelent.

```python
N_1, N_2, N_3, N_4, N_5, N_6, N_7, N_8, N_9, N_10, N_11 = sp.symbols('N_1, N_2, N_3, N_4, N_5, N_6, N_7, N_8, N_9, N_10, N_11')

# A ferde rudak vízszintessel bezárt szöge
alpha = sp.rad(45) # rad
```

##### $B$ csukló

<img src="img/03.jpg" alt="Feladat" width=200px>

```python
# x irányú erő egyensúly
eq1 = -N_11 - N_6*sp.cos(alpha)
display(Math(rf'\sum F_x = 0 \quad \longrightarrow \quad {latex(eq1)} = 0'))

# y irányú erő egyensúly
eq2 = B_ysol + N_6*sp.sin(alpha)
display(Math(rf'\sum F_y = 0 \quad \longrightarrow \quad {latex(eq2)} = 0'))
```

```python
sol = sp.solve([eq1, eq2], [N_6, N_11])

N_6sol = sol[N_6]
display(Math(rf'N_6 = {latex(round(N_6sol, 2))} ~ \mathrm{{N}}'))

N_11sol = sol[N_11]
display(Math(rf'N_{{11}} = {latex(round(N_11sol, 2))} ~ \mathrm{{N}}'))
```

##### $G$ csukló

<img src="img/04.jpg" alt="Feladat" width=200px>

```python
# x irányú erő egyensúly
eq1 = -N_10 + N_11sol
display(Math(rf'\sum F_x = 0 \quad \longrightarrow \quad {latex(eq1)} = 0'))

# y irányú erő egyensúly
eq2 = N_5
display(Math(rf'\sum F_y = 0 \quad \longrightarrow \quad {latex(eq2)} = 0'))
```

```python
sol = sp.solve([eq1, eq2], [N_5, N_10])

N_5sol = sol[N_5]
display(Math(rf'N_5 = {latex(round(N_5sol, 2))} ~ \mathrm{{N}}'))

N_10sol = sol[N_10]
display(Math(rf'N_{{10}} = {latex(round(N_10sol, 2))} ~ \mathrm{{N}}'))
```

Az 5. rúdban nem ébred erő, tehát az 5. rúd vakrúd.

##### $D$ csukló

<img src="img/05.jpg" alt="Feladat" width=220px>

```python
# x irányú erő egyensúly
eq1 = -N_7 - N_4*sp.cos(alpha) + N_6sol*sp.cos(alpha) - F_2
display(Math(rf'\sum F_x = 0 \quad \longrightarrow \quad {latex(eq1)} = 0'))

# y irányú erő egyensúly
eq2 = -N_4*sp.sin(alpha) - N_5sol - N_6sol*sp.sin(alpha)
display(Math(rf'\sum F_y = 0 \quad \longrightarrow \quad {latex(eq2)} = 0'))
```

```python
sol = sp.solve([eq1, eq2], [N_4, N_7])

N_4sol = sol[N_4]
display(Math(rf'N_4 = {latex(round(N_4sol, 2))} ~ \mathrm{{N}}'))

N_7sol = sol[N_7]
display(Math(rf'N_7 = {latex(round(N_7sol, 2))} ~ \mathrm{{N}}'))
```

##### $A$ csukló

<img src="img/06.jpg" alt="Feladat" width=200px>

```python
# x irányú erő egyensúly
eq1 = A_xsol + N_1*sp.cos(alpha) + N_8
display(Math(rf'\sum F_x = 0 \quad \longrightarrow \quad {latex(eq1)} = 0'))

# y irányú erő egyensúly
eq2 = A_ysol + N_1*sp.sin(alpha)
display(Math(rf'\sum F_y = 0 \quad \longrightarrow \quad {latex(eq2)} = 0'))
```

```python
sol = sp.solve([eq1, eq2], [N_1, N_8])

N_1sol = sol[N_1]
display(Math(rf'N_1 = {latex(round(N_1sol, 2))} ~ \mathrm{{N}}'))

N_8sol = sol[N_8]
display(Math(rf'N_8 = {latex(round(N_8sol, 2))} ~ \mathrm{{N}}'))
```

##### $E$ csukló

<img src="img/07.jpg" alt="Feladat" width=200px>

```python
# x irányú erő egyensúly
eq1 = -N_8sol + N_9
display(Math(rf'\sum F_x = 0 \quad \longrightarrow \quad {latex(eq1)} = 0'))

# y irányú erő egyensúly
eq2 = N_2
display(Math(rf'\sum F_y = 0 \quad \longrightarrow \quad {latex(eq2)} = 0'))
```

```python
sol = sp.solve([eq1, eq2], [N_2, N_9])

N_2sol = sol[N_2]
display(Math(rf'N_2 = {latex(round(N_2sol, 2))} ~ \mathrm{{N}}'))

N_9sol = sol[N_9]
display(Math(rf'N_9 = {latex(round(N_9sol, 2))} ~ \mathrm{{N}}'))
```

A 2. rúdban nem ébred erő, tehát a 2. rúd vakrúd.

##### $F$ csukló

<img src="img/08.jpg" alt="Feladat" width=220px>

```python
# x irányú erő egyensúly
eq1 = -N_9sol - N_3*sp.cos(alpha) + N_4sol*sp.cos(alpha) + N_10sol
display(Math(rf'\sum F_x = 0 \quad \longrightarrow \quad {latex(eq1)} = 0'))

# y irányú erő egyensúly
eq2 = N_3*sp.sin(alpha) + N_4sol*sp.sin(alpha)
display(Math(rf'\sum F_y = 0 \quad \longrightarrow \quad {latex(eq2)} = 0'))
```

```python
# Az y irányú egyenletből számolható N_3
N_3sol = sp.solve(eq2, N_3)[0]
display(Math(rf'N_3 = {latex(round(N_3sol, 2))} ~ \mathrm{{N}}'))
```

##### Ellenőrzés

Minden rúderő ismert.

11 rúd → 11 ismeretlen, 7 csukló → 7·2 = 14 egyenlet!

A 3 reakcióerőt az egész szerkezetre felírt egyensúlyi egyenletekből határoztuk meg, így a még fel nem használt egyenletek ($F$ csuklóból +1 egyenlet és $C$ csuklóból +2 egyenlet) ellenőrzésre használhatók.

```python
# F csukló, x irányú erő egyensúly
chk1 = eq1.subs(N_3, N_3sol)
display(Math(rf'F \text{{ csukló: }} \quad \sum F_x = {latex(sp.simplify(chk1))}'))

# C csukló, x irányú erő egyensúly
chk2 = -N_1sol*sp.cos(alpha) + N_3sol*sp.cos(alpha) + N_7sol
display(Math(rf'C \text{{ csukló: }} \quad \sum F_x = {latex(sp.simplify(chk2))}'))

# C csukló, y irányú erő egyensúly
chk3 = -N_1sol*sp.sin(alpha) - N_2sol - N_3sol*sp.sin(alpha) - F_1
display(Math(rf'C \text{{ csukló: }} \quad \sum F_y = {latex(sp.simplify(chk3))}'))
```

Látható, hogy az egyenletek teljesülnek.

##### A rúderők összefoglalása

```python
# Az egyes rúderők listába gyűjtve
Ni = [N_1sol, N_2sol, N_3sol, N_4sol, N_5sol, N_6sol, N_7sol, N_8sol, N_9sol, N_10sol, N_11sol]

for i in range(len(Ni)):
    # A rúd állapota a rúderő előjele alapján
    if abs(Ni[i]) < 1e-6:
        allapot = 'vakrúd'
    elif Ni[i] > 0:
        allapot = 'húzott'
    else:
        allapot = 'nyomott'
    display(Math(rf'N_{{{i+1}}} = {latex(round(Ni[i], 2))} ~ \mathrm{{N}} \quad \text{{({allapot})}}'))
```
