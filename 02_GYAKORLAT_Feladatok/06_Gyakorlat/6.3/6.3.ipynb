##### Csomagok importálása

```python
import numpy as np # Numerikus számítások
import matplotlib.pyplot as plt # Ábrázolás
plt.rcParams['mathtext.fontset'] = 'cm' # Betűtípus az ábrákon
plt.rcParams['font.family'] = 'serif' # Betűtípus az ábrákon
plt.rcParams['font.serif'] = ['cmr10'] # Betűtípus az ábrákon
plt.rcParams['axes.formatter.use_mathtext'] = True # Betűtípus az ábrákon
plt.rcParams['axes.unicode_minus'] = False # Betűtípus az ábrákon
from IPython.display import display, Math, Latex # Szép kimenet Jupyter notebookban
import sympy as sp # Szimbólikus számítások
from sympy import latex # Szép kimenet Jupyter notebookban
sp.init_printing() # Szép kimenet Jupyter notebookban
```

##### Adatok

```python
a = 1 # m
b = 2 # m
c = 3 # m

F = 1000 # N
```

##### Szabadtest ábra

<img src="img/02.jpg" alt="Feladat" width=280px>

A reakciókra 4 ismeretlen ($A_x$, $A_y$, $B_x$, $B_y$) adódik, de csak 3 egyensúlyi egyenletünk van, így a reakcióerők az egész szerkezetre felírt egyensúlyi egyenletekből nem határozhatók meg.

→ Indulás az $E$ pontból (max. 2 ismeretlen).

##### A rúderők és a reakciók

* Mindig olyan csuklót válasszunk, ahol max. 2 ismeretlen van!

* Az ismeretlen rúderőket úgy vegyük fel, mintha a rúd húzott lenne! Így a pozitív rúderő húzott, a negatív rúderő nyomott rudat jelent.

```python
N_1, N_2, N_3, N_4, N_5, N_6 = sp.symbols('N_1, N_2, N_3, N_4, N_5, N_6')
A_x, A_y, B_x, B_y = sp.symbols('A_x, A_y, B_x, B_y')
```

##### $E$ csukló

<img src="img/03.jpg" alt="Feladat" width=200px>

```python
# Az 5. rúd függőlegessel bezárt szöge a geometria alapján
alpha = sp.atan(a/c)
display(Math(rf'\alpha = {latex(sp.deg(alpha).evalf(5))}^\circ'))
```

```python
# x irányú erő egyensúly
eq1 = F - N_5*sp.sin(alpha)
display(Math(rf'\sum F_x = 0 \quad \longrightarrow \quad {latex(eq1.evalf(4))} = 0'))

# y irányú erő egyensúly
eq2 = -N_5*sp.cos(alpha) - N_6
display(Math(rf'\sum F_y = 0 \quad \longrightarrow \quad {latex(eq2.evalf(4))} = 0'))
```

```python
sol = sp.solve([eq1, eq2], [N_5, N_6])

N_5sol = sol[N_5]
display(Math(rf'N_5 = {latex(round(N_5sol, 2))} ~ \mathrm{{N}}'))

N_6sol = sol[N_6]
display(Math(rf'N_6 = {latex(round(N_6sol, 2))} ~ \mathrm{{N}}'))
```

##### $D$ csukló

<img src="img/04.jpg" alt="Feladat" width=200px>

```python
# x irányú erő egyensúly
eq1 = -N_4
display(Math(rf'\sum F_x = 0 \quad \longrightarrow \quad {latex(eq1.evalf(4))} = 0'))

# y irányú erő egyensúly
eq2 = N_6sol - N_3
display(Math(rf'\sum F_y = 0 \quad \longrightarrow \quad {latex(eq2.evalf(4))} = 0'))
```

```python
sol = sp.solve([eq1, eq2], [N_3, N_4])

N_3sol = sol[N_3]
display(Math(rf'N_3 = {latex(round(N_3sol, 2))} ~ \mathrm{{N}}'))

N_4sol = sol[N_4]
display(Math(rf'N_4 = {latex(round(N_4sol, 2))} ~ \mathrm{{N}}'))
```

A 4. rúdban nem ébred erő, tehát a 4. rúd vakrúd.

##### $C$ csukló

<img src="img/05.jpg" alt="Feladat" width=200px>

```python
# Az 5. rúd vízszintessel bezárt szöge a geometria alapján
beta = sp.atan(c/a)
display(Math(rf'\beta = {latex(sp.deg(beta).evalf(5))}^\circ'))

# A 2. rúd vízszintessel bezárt szöge a geometria alapján
gamma = sp.atan(b/a)
display(Math(rf'\gamma = {latex(sp.deg(gamma).evalf(5))}^\circ'))
```

```python
# x irányú erő egyensúly
eq1 = N_5sol*sp.cos(beta) + N_4sol + N_2*sp.cos(gamma)
display(Math(rf'\sum F_x = 0 \quad \longrightarrow \quad {latex(eq1.evalf(4))} = 0'))

# y irányú erő egyensúly
eq2 = -N_1 + N_5sol*sp.sin(beta) - N_2*sp.sin(gamma)
display(Math(rf'\sum F_y = 0 \quad \longrightarrow \quad {latex(eq2.evalf(4))} = 0'))
```

```python
sol = sp.solve([eq1, eq2], [N_1, N_2])

N_1sol = sol[N_1]
display(Math(rf'N_1 = {latex(round(N_1sol, 2))} ~ \mathrm{{N}}'))

N_2sol = sol[N_2]
display(Math(rf'N_2 = {latex(round(N_2sol, 2))} ~ \mathrm{{N}}'))
```

##### $A$ csukló

<img src="img/06.jpg" alt="Feladat" width=200px>

```python
# x irányú erő egyensúly
eq1 = A_x
display(Math(rf'\sum F_x = 0 \quad \longrightarrow \quad {latex(eq1.evalf(4))} = 0'))

# y irányú erő egyensúly
eq2 = A_y + N_1sol
display(Math(rf'\sum F_y = 0 \quad \longrightarrow \quad {latex(eq2.evalf(4))} = 0'))
```

```python
sol = sp.solve([eq1, eq2], [A_x, A_y])

A_xsol = sol[A_x]
display(Math(rf'A_x = {latex(round(A_xsol, 2))} ~ \mathrm{{N}}'))

A_ysol = sol[A_y]
display(Math(rf'A_y = {latex(round(A_ysol, 2))} ~ \mathrm{{N}}'))
```

##### $B$ csukló

<img src="img/07.jpg" alt="Feladat" width=200px>

```python
# A 2. rúd vízszintessel bezárt szöge a geometria alapján
delta = sp.atan(b/a)
display(Math(rf'\delta = {latex(sp.deg(delta).evalf(5))}^\circ'))
```

```python
# x irányú erő egyensúly
eq1 = B_x - N_2sol*sp.cos(delta)
display(Math(rf'\sum F_x = 0 \quad \longrightarrow \quad {latex(eq1.evalf(4))} = 0'))

# y irányú erő egyensúly
eq2 = N_2sol*sp.sin(delta) + N_3sol + B_y
display(Math(rf'\sum F_y = 0 \quad \longrightarrow \quad {latex(eq2.evalf(4))} = 0'))
```

```python
sol = sp.solve([eq1, eq2], [B_x, B_y])

B_xsol = sol[B_x]
display(Math(rf'B_x = {latex(round(B_xsol, 2))} ~ \mathrm{{N}}'))

B_ysol = sol[B_y]
display(Math(rf'B_y = {latex(round(B_ysol, 2))} ~ \mathrm{{N}}'))
```

##### Ellenőrzés

A reakcióerők ismeretében az egész szerkezetre felírt egyensúlyi egyenletek ellenőrzésre használhatók.

```python
# x irányú erő egyensúly
chk1 = A_xsol + B_xsol + F
display(Math(rf'\sum F_x = {latex(round(chk1, 2))}'))

# y irányú erő egyensúly
chk2 = A_ysol + B_ysol
display(Math(rf'\sum F_y = {latex(round(chk2, 2))}'))

# z irányú nyomatéki egyensúly
chk3 = -F*(b + c) + B_ysol*a
display(Math(rf'\sum M_A = {latex(round(chk3, 2))}'))
```

Látható, hogy az egyenletek teljesülnek.

##### A rúderők összefoglalása

```python
# Az egyes rúderők listába gyűjtve
Ni = [N_1sol, N_2sol, N_3sol, N_4sol, N_5sol, N_6sol]

for i in range(len(Ni)):
    # A rúd állapota a rúderő előjele alapján
    if abs(Ni[i]) < 1e-6:
        allapot = 'vakrúd'
    elif Ni[i] > 0:
        allapot = 'húzott'
    else:
        allapot = 'nyomott'
    display(Math(rf'N_{{{i+1}}} = {latex(round(Ni[i], 2))} ~ \mathrm{{N}} \quad \text{{({allapot})}}'))
```
