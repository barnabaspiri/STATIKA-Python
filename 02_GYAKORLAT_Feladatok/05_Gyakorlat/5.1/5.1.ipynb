##### Csomagok importálása

```python
import numpy as np # Numerikus számítások
import matplotlib.pyplot as plt # Ábrázolás
plt.rcParams['mathtext.fontset'] = 'cm' # Betűtípus az ábrákon
plt.rcParams['font.family'] = 'serif' # Betűtípus az ábrákon
plt.rcParams['font.serif'] = ['cmr10'] # Betűtípus az ábrákon
plt.rcParams['axes.formatter.use_mathtext'] = True # Betűtípus az ábrákon
plt.rcParams['axes.unicode_minus'] = False # Betűtípus az ábrákon
from IPython.display import display, Math, Latex # Szép kimenet Jupyter notebookban
import sympy as sp # Szimbólikus számítások
from sympy import latex # Szép kimenet Jupyter notebookban
sp.init_printing() # Szép kimenet Jupyter notebookban
```

##### Adatok

```python
h = 1 # m
r = 0.5 # m
```

##### a) Vonalakból áll

Az alábbi ábra szerint 3 részre bontható: 2 egyenes és 1 félkörív

<img src="img/02.jpg" alt="Feladat" width=210px>

Az egyes részek tulajdonságai

```python
# 1. egyenes
r_s1 = sp.Matrix([-r/2, h/2])
l1 = sp.sqrt(h**2 + r**2)

# 2. egyenes
r_s2 = sp.Matrix([r/2, h/2])
l2 = sp.sqrt(h**2 + r**2)

# 3. félkörív
r_s3 = sp.Matrix([0, -2*r/sp.pi])
l3 = sp.pi * r
```

A súlyponti koordináták

```python
# Az egyes elemek tulajdonságai listákba gyűjtve
r_si = [r_s1, r_s2, r_s3]
li = [l1, l2, l3]

x_s = sum(r_si[i][0] * li[i] for i in range(len(li))) / sum(li)
display(Math(rf'x_s =  {latex(x_s)} ~\mathrm{{m}}'))

y_s = sum(r_si[i][1] * li[i] for i in range(len(li))) / sum(li)
display(Math(rf'y_s =  {latex(y_s.evalf(4))} ~\mathrm{{m}}'))
```

##### b) Sík felületekből áll

Az alábbi ábra szerint 2 részre bontható: 1 háromszög és 1 félkör

<img src="img/03.jpg" alt="Feladat" width=210px>

Az egyes részek tulajdonságai

```python
# 1. háromszög
r_s1 = sp.Matrix([0, h/3])
A1 = r*h

# 2. félkör
r_s2 = sp.Matrix([0, (-4*r)/(3*sp.pi)])
A2 = r**2*sp.pi/2
```

A súlyponti koordináták

```python
# Az egyes elemek tulajdonságai listákba gyűjtve
r_si = [r_s1, r_s2]
Ai = [A1, A2]

x_s = sum(r_si[i][0] * Ai[i] for i in range(len(Ai))) / sum(Ai)
display(Math(rf'x_s =  {latex(x_s)} ~\mathrm{{m}}'))

y_s = sum(r_si[i][1] * Ai[i] for i in range(len(Ai))) / sum(Ai)
display(Math(rf'y_s =  {latex(y_s.evalf(4))} ~\mathrm{{m}}'))
```

##### c) Testekből áll

Az alábbi ábra szerint 2 részre bontható: 1 kúp és 1 félgömb

<img src="img/04.jpg" alt="Feladat" width=210px>

Az egyes részek tulajdonságai

```python
# 1. háromszög
r_s1 = sp.Matrix([0, h/4])
V1 = r**2*sp.pi*h/3

# 2. félkör
r_s2 = sp.Matrix([0, (-3*r)/(8)])
V2 = 1/2*4*r**3*sp.pi/3
```

A súlyponti koordináták

```python
# Az egyes elemek tulajdonságai listákba gyűjtve
r_si = [r_s1, r_s2]
Vi = [V1, V2]

x_s = sum(r_si[i][0] * Vi[i] for i in range(len(Vi))) / sum(Vi)
display(Math(rf'x_s =  {latex(x_s)} ~\mathrm{{m}}'))

y_s = sum(r_si[i][1] * Vi[i] for i in range(len(Vi))) / sum(Vi)
display(Math(rf'y_s =  {latex(y_s.evalf(4))} ~\mathrm{{m}}'))
```
