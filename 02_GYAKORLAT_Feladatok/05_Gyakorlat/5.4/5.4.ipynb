##### Feladat

Az ábrán látható sorompó az $a$ élhosszúságú kockából és az $L$ hosszúságú négyzet keresztmetszetű rúdból áll. A fa rúd keresztmetszeti mérete $b$. Mekkora legyen $L$, hogy az $A$ pontban alátámasztva egyensúlyban legyen?

##### Csomagok importálása

```python
import numpy as np # Numerikus számítások
import matplotlib.pyplot as plt # Ábrázolás
plt.rcParams['mathtext.fontset'] = 'cm' # Betűtípus az ábrákon
plt.rcParams['font.family'] = 'serif' # Betűtípus az ábrákon
plt.rcParams['font.serif'] = ['cmr10'] # Betűtípus az ábrákon
plt.rcParams['axes.formatter.use_mathtext'] = True # Betűtípus az ábrákon
plt.rcParams['axes.unicode_minus'] = False # Betűtípus az ábrákon
from IPython.display import display, Math, Latex # Szép kimenet Jupyter notebookban
import sympy as sp # Szimbólikus számítások
from sympy import latex # Szép kimenet Jupyter notebookban
sp.init_printing() # Szép kimenet Jupyter notebookban
```

##### Adatok

```python
a = 0.12 # m
b = 0.040 # m

rho_a = 7800 # kg/m^3
rho_f = 600 # kg/m^3
```

```python
L = sp.Symbol('L', nonzero=True, positive=True)
```

##### A statikai nyomaték az $A$ pontra zérus

```python
# Acél kocka tulajdonságai
x_a = -a
m_a = a**3 * rho_a

# Fa rúd tulajdonságai
x_f = L/2 - a/2
m_f = b**2 * L * rho_f

# Statikai nyomaték az A pontra
eq = sp.expand(m_a*x_a + m_f*x_f)
display(Math(rf'S_A = x_a\cdot m_a = x_f \cdot m_f = {latex(eq)}'))
```

##### Az egyenlet megoldása $L$-re

```python
Lsol = sp.solve(eq, L)[0]
display(Math(rf'L = {latex(Lsol.evalf(5))} ~ \mathrm{{m}}'))
```
