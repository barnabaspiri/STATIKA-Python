##### Csomagok importálása

```python
import numpy as np # Numerikus számítások
import matplotlib.pyplot as plt # Ábrázolás
plt.rcParams['mathtext.fontset'] = 'cm' # Betűtípus az ábrákon
plt.rcParams['font.family'] = 'serif' # Betűtípus az ábrákon
plt.rcParams['font.serif'] = ['cmr10'] # Betűtípus az ábrákon
plt.rcParams['axes.formatter.use_mathtext'] = True # Betűtípus az ábrákon
plt.rcParams['axes.unicode_minus'] = False # Betűtípus az ábrákon
from IPython.display import display, Math, Latex # Szép kimenet Jupyter notebookban
import sympy as sp # Szimbólikus számítások
from sympy import latex # Szép kimenet Jupyter notebookban
sp.init_printing() # Szép kimenet Jupyter notebookban
```

##### Adatok

```python
a = 1 # m

F_1 = 300 # N
F_2 = 400 # N
```

##### Csuklós szerkezetek

* A szerkezet elemei csuklósan kapcsolódnak.

* Nem csak a csuklóban van terhelés.

* 2 db csukló → 4 ismeretlen ($A_x$, $A_y$, $B_x$, $B_y$), de csak 3 egyensúlyi egyenletünk van, így a reakcióerők az egész szerkezetre felírt egyensúlyi egyenletekből nem határozhatók meg.

Megoldási módszerek: 1. részekre bontás, 2. szuperpozíció.

##### Számítással (részekre bontás)

A szerkezetet a $C$ csuklónál 2 részre bontjuk:

* 6 db ismeretlen: $A_x$, $A_y$, $B_x$, $B_y$, $C_x$, $C_y$

* 2·3 egyensúlyi egyenlet

→ megoldás

##### Szabadtest ábra

<img src="img/02.jpg" alt="Feladat" width=600px>

##### A reakciók

```python
A_x, A_y, B_x, B_y, C_x, C_y = sp.symbols('A_x, A_y, B_x, B_y, C_x, C_y')
```

##### Egyensúlyi egyenletek az $AC$ rúdra

```python
# x irányú erő egyensúly
eq1 = A_x + F_1 + C_x
display(Math(rf'\sum F_x = 0 \quad \longrightarrow \quad {latex(eq1)} = 0 \qquad (1)'))

# y irányú erő egyensúly
eq2 = A_y + C_y
display(Math(rf'\sum F_y = 0 \quad \longrightarrow \quad {latex(eq2)} = 0 \qquad (2)'))

# z irányú nyomatéki egyensúly a C pontra
eq3 = F_1*2*a + A_x*3*a - A_y*a
display(Math(rf'\sum M_C = 0 \quad \longrightarrow \quad {latex(eq3)} = 0 \qquad (3)'))
```

##### Egyensúlyi egyenletek a $BC$ rúdra

```python
# x irányú erő egyensúly
eq4 = -C_x + B_x
display(Math(rf'\sum F_x = 0 \quad \longrightarrow \quad {latex(eq4)} = 0 \qquad (4)'))

# y irányú erő egyensúly
eq5 = -C_y - F_2 + B_y
display(Math(rf'\sum F_y = 0 \quad \longrightarrow \quad {latex(eq5)} = 0 \qquad (5)'))

# z irányú nyomatéki egyensúly a C pontra
eq6 = -F_2*a + B_y*3*a + B_x*3*a
display(Math(rf'\sum M_C = 0 \quad \longrightarrow \quad {latex(eq6)} = 0 \qquad (6)'))
```

##### Egyensúlyi egyenletek megoldása

6 ismeretlen és 6 egyenlet → megoldás. Python segítségével az egyenletrendszer közvetlenül megoldható:

```python
sol = sp.solve([eq1, eq2, eq3, eq4, eq5, eq6], [A_x, A_y, B_x, B_y, C_x, C_y])

A_xsol = sol[A_x]
display(Math(rf'A_x = {latex(round(A_xsol, 2))} ~ \mathrm{{N}}'))

A_ysol = sol[A_y]
display(Math(rf'A_y = {latex(round(A_ysol, 2))} ~ \mathrm{{N}}'))

B_xsol = sol[B_x]
display(Math(rf'B_x = {latex(round(B_xsol, 2))} ~ \mathrm{{N}}'))

B_ysol = sol[B_y]
display(Math(rf'B_y = {latex(round(B_ysol, 2))} ~ \mathrm{{N}}'))

C_xsol = sol[C_x]
display(Math(rf'C_x = {latex(round(C_xsol, 2))} ~ \mathrm{{N}}'))

C_ysol = sol[C_y]
display(Math(rf'C_y = {latex(round(C_ysol, 2))} ~ \mathrm{{N}}'))
```

##### Egyensúly a teljes szerkezetre

Kézi számolásnál a 6 egyenletből álló egyenletrendszer megoldása elég hosszú is lehet. Egyszerűbb, ha a teljes szerkezetre felírt egyensúlyi egyenleteket is felhasználjuk:

<img src="img/03.jpg" alt="Feladat" width=300px>

```python
# x irányú erő egyensúly
eq7 = A_x + F_1 + B_x
display(Math(rf'\sum F_x = 0 \quad \longrightarrow \quad {latex(eq7)} = 0 \qquad (7)'))

# y irányú erő egyensúly
eq8 = A_y - F_2 + B_y
display(Math(rf'\sum F_y = 0 \quad \longrightarrow \quad {latex(eq8)} = 0 \qquad (8)'))

# z irányú nyomatéki egyensúly az A pontra
eq9 = -F_1*a - F_2*2*a + B_y*4*a
display(Math(rf'\sum M_A = 0 \quad \longrightarrow \quad {latex(eq9)} = 0 \qquad (9)'))
```

Ezekből és a részekre felírt egyenletekből a reakciók egyesével kifejezhetők:

```python
# (9)-ből
B_ysol = sp.solve(eq9, B_y)[0]
display(Math(rf'B_y = {latex(round(B_ysol, 2))} ~ \mathrm{{N}}'))

# (8)-ból
A_ysol = sp.solve(eq8.subs(B_y, B_ysol), A_y)[0]
display(Math(rf'A_y = {latex(round(A_ysol, 2))} ~ \mathrm{{N}}'))

# (6)-ból
B_xsol = sp.solve(eq6.subs(B_y, B_ysol), B_x)[0]
display(Math(rf'B_x = {latex(round(B_xsol, 2))} ~ \mathrm{{N}}'))

# (3)-ból
A_xsol = sp.solve(eq3.subs(A_y, A_ysol), A_x)[0]
display(Math(rf'A_x = {latex(round(A_xsol, 2))} ~ \mathrm{{N}}'))

# (2)-ből
C_ysol = sp.solve(eq2.subs(A_y, A_ysol), C_y)[0]
display(Math(rf'C_y = {latex(round(C_ysol, 2))} ~ \mathrm{{N}}'))

# (4)-ből
C_xsol = sp.solve(eq4.subs(B_x, B_xsol), C_x)[0]
display(Math(rf'C_x = {latex(round(C_xsol, 2))} ~ \mathrm{{N}}'))
```

Látható, hogy ugyanazt az eredményt kaptuk.

##### Ellenőrzés

Az (1), (5) és (7) egyenletet még nem használtuk fel, így ezek ellenőrzésre használhatók.

```python
# AC rúd, x irányú erő egyensúly (1)
chk1 = eq1.subs({A_x: A_xsol, C_x: C_xsol})
display(Math(rf'AC \text{{ rúd: }} \quad \sum F_x = {latex(round(chk1, 2))}'))

# BC rúd, y irányú erő egyensúly (5)
chk2 = eq5.subs({C_y: C_ysol, B_y: B_ysol})
display(Math(rf'BC \text{{ rúd: }} \quad \sum F_y = {latex(round(chk2, 2))}'))

# Teljes szerkezet, x irányú erő egyensúly (7)
chk3 = eq7.subs({A_x: A_xsol, B_x: B_xsol})
display(Math(rf'\text{{Teljes szerkezet: }} \quad \sum F_x = {latex(round(chk3, 2))}'))
```

Látható, hogy az egyenletek teljesülnek.

##### Szerkesztéssel (szuperpozíciót felhasználva)

* A külső terhelést több részben vizsgáljuk: az 1. részben csak az 1. ($AC$) rúdra ható erők, a 2. részben csak a 2. ($BC$) rúdra ható erők hatnak.

* Szuperpozíció: a teljes reakciókat a külön vizsgált részek reakcióerőinek összege adja:

$$A_x = A_x' + A_x'' \qquad B_x = B_x' + B_x''$$

$$A_y = A_y' + A_y'' \qquad B_y = B_y' + B_y''$$

* Ezt azért tehetjük meg, mert az egyensúlyi egyenletek lineáris rendszert alkotnak → ha a külső erőt „n” szeresére növelem, a reakciók „n” szeresére növekednek.

* Szerkesztésnél fontos a lépték!

```python
x, y = sp.symbols('x, y')

# A csuklók helyvektorai
r_A = sp.Matrix([0, 0])
r_B = sp.Matrix([4*a, 0])
r_C = sp.Matrix([a, 3*a])
```

```python
def nyil(ax, kezdo, vektor, szin, felirat, oldal=1):
    # Nyíl rajzolása a kezdőpontból
    ax.annotate('', xy=kezdo + vektor, xytext=kezdo,
                arrowprops=dict(arrowstyle='-|>', color=szin, lw=1.5, shrinkA=0, shrinkB=0))
    # Felirat a nyíl közepénél, a nyílra merőlegesen eltolva (oldal = 1: balra, oldal = -1: jobbra)
    n = oldal * np.array([-vektor[1], vektor[0]]) / np.linalg.norm(vektor)
    ax.annotate(felirat, xy=kezdo + vektor/2, xytext=14*n, textcoords='offset points',
                ha='center', va='center', color=szin, fontsize=12)
    # A tengelyhatárok frissítése
    ax.update_datalim([kezdo, kezdo + vektor])
    return kezdo + vektor
```

##### 1. rész: csak az $F_1$ erő hat

<img src="img/04.jpg" alt="Feladat" width=350px>

Ismeretlenek: $A_x'$, $A_y'$, $B_x'$, $B_y'$, de $B_x'$ és $B_y'$ nem függetlenek, mivel a 2. ($BC$) rúd terheletlen, így a $\mathbf{B}'$ erő rúdirányú → 3 egyensúlyi egyenlet → megoldás.

A szerkesztés lépései:

1. $e_B'$ és $e_{F_1}$ hatásvonal ismert
2. metszéspontjuk: $P_1$
3. $e_A'$ hatásvonal átmegy $P_1$-en
4. vektorháromszög

```python
# 1.) e_B' hatásvonal: a BC rúd egyenese
e_B1 = sp.Eq(y, -(x - 4*a))
display(Math(rf"e_B': \quad {latex(e_B1)}"))

# 1.) e_F1 hatásvonal: vízszintes egyenes y = a magasságban
e_F1 = sp.Eq(y, a)
display(Math(rf'e_{{F_1}}: \quad {latex(e_F1)}'))

# 2.) A hatásvonalak metszéspontja
sol = sp.solve([e_B1, e_F1], [x, y])
r_P1 = sp.Matrix([sol[x], sol[y]])
display(Math(rf'P_1 = {latex(r_P1)} ~ \mathrm{{m}}'))

# 3.) e_A' hatásvonal: átmegy az A ponton és a P_1 ponton
e_A1 = sp.Eq(y, r_A[1] + (r_P1[1] - r_A[1]) / (r_P1[0] - r_A[0]) * (x - r_A[0]))
display(Math(rf"e_A': \quad {latex(e_A1)}"))
```

```python
# 4.) Vektorháromszög: F_1 + B' + A' = 0
A_1, B_1 = sp.symbols("A', B'") # Az erők előjeles nagysága

# A hatásvonalak irányába mutató egységvektorok
n_A1 = (r_P1 - r_A) / (r_P1 - r_A).norm()
n_B1 = (r_C - r_B) / (r_C - r_B).norm()

F1_vec = sp.Matrix([F_1, 0])

sol = sp.solve(list(F1_vec + B_1*n_B1 + A_1*n_A1), [A_1, B_1])

A1_vec = sol[A_1]*n_A1
display(Math(rf"\mathbf{{A}}' = {latex(A1_vec.evalf(5))} ~ \mathrm{{N}}"))

B1_vec = sol[B_1]*n_B1
display(Math(rf"\mathbf{{B}}' = {latex(B1_vec.evalf(5))} ~ \mathrm{{N}}"))
```

A vektorháromszög léptékhelyesen:

```python
# Az erővektorok numpy tömbként
F1_v = np.array(F1_vec, dtype=float).flatten()
A1_v = np.array(A1_vec, dtype=float).flatten()
B1_v = np.array(B1_vec, dtype=float).flatten()

fig, ax = plt.subplots(figsize=(10/2.54, 8/2.54))
p = nyil(ax, np.zeros(2), F1_v, 'red', r'$\mathbf{F}_1$', -1)
p = nyil(ax, p, B1_v, 'blue', r"$\mathbf{B}'$", -1)
p = nyil(ax, p, A1_v, 'darkorange', r"$\mathbf{A}'$", -1)
ax.set_aspect('equal', adjustable='datalim')
ax.margins(0.2)
ax.autoscale_view()
ax.set_xlabel(r'$F_x ~ [\mathrm{N}]$')
ax.set_ylabel(r'$F_y ~ [\mathrm{N}]$')
ax.grid(True, alpha=0.5)
plt.tight_layout()
plt.show()
```

##### 2. rész: csak az $F_2$ erő hat

<img src="img/05.jpg" alt="Feladat" width=330px>

Ismeretlenek: $A_x''$, $A_y''$, $B_x''$, $B_y''$, de $A_x''$ és $A_y''$ nem függetlenek, mivel az 1. ($AC$) rúd terheletlen, így az $\mathbf{A}''$ erő rúdirányú → 3 egyensúlyi egyenlet → megoldás.

A szerkesztés lépései:

1. $e_A''$ és $e_{F_2}$ hatásvonal ismert
2. metszéspontjuk: $P_2$
3. $e_B''$ hatásvonal átmegy $P_2$-n
4. vektorháromszög

```python
# 1.) e_A'' hatásvonal: az AC rúd egyenese
e_A2 = sp.Eq(y, 3*x)
display(Math(rf"e_A'': \quad {latex(e_A2)}"))

# 1.) e_F2 hatásvonal: függőleges egyenes x = 2a helyen
e_F2 = sp.Eq(x, 2*a)
display(Math(rf'e_{{F_2}}: \quad {latex(e_F2)}'))

# 2.) A hatásvonalak metszéspontja
sol = sp.solve([e_A2, e_F2], [x, y])
r_P2 = sp.Matrix([sol[x], sol[y]])
display(Math(rf'P_2 = {latex(r_P2)} ~ \mathrm{{m}}'))

# 3.) e_B'' hatásvonal: átmegy a B ponton és a P_2 ponton
e_B2 = sp.Eq(y, r_B[1] + (r_P2[1] - r_B[1]) / (r_P2[0] - r_B[0]) * (x - r_B[0]))
display(Math(rf"e_B'': \quad {latex(e_B2)}"))
```

```python
# 4.) Vektorháromszög: F_2 + A'' + B'' = 0
A_2, B_2 = sp.symbols("A'', B''") # Az erők előjeles nagysága

# A hatásvonalak irányába mutató egységvektorok
n_A2 = (r_C - r_A) / (r_C - r_A).norm()
n_B2 = (r_P2 - r_B) / (r_P2 - r_B).norm()

F2_vec = sp.Matrix([0, -F_2])

sol = sp.solve(list(F2_vec + A_2*n_A2 + B_2*n_B2), [A_2, B_2])

A2_vec = sol[A_2]*n_A2
display(Math(rf"\mathbf{{A}}'' = {latex(A2_vec.evalf(5))} ~ \mathrm{{N}}"))

B2_vec = sol[B_2]*n_B2
display(Math(rf"\mathbf{{B}}'' = {latex(B2_vec.evalf(5))} ~ \mathrm{{N}}"))
```

A vektorháromszög léptékhelyesen:

```python
# Az erővektorok numpy tömbként
F2_v = np.array(F2_vec, dtype=float).flatten()
A2_v = np.array(A2_vec, dtype=float).flatten()
B2_v = np.array(B2_vec, dtype=float).flatten()

fig, ax = plt.subplots(figsize=(10/2.54, 8/2.54))
p = nyil(ax, np.zeros(2), F2_v, 'red', r'$\mathbf{F}_2$', -1)
p = nyil(ax, p, A2_v, 'darkorange', r"$\mathbf{A}''$", -1)
p = nyil(ax, p, B2_v, 'blue', r"$\mathbf{B}''$", -1)
ax.set_aspect('equal', adjustable='datalim')
ax.margins(0.2)
ax.autoscale_view()
ax.set_xlabel(r'$F_x ~ [\mathrm{N}]$')
ax.set_ylabel(r'$F_y ~ [\mathrm{N}]$')
ax.grid(True, alpha=0.5)
plt.tight_layout()
plt.show()
```

##### Összegezve

A teljes reakciókat a két rész reakcióerőinek összege adja:

```python
A_vec = A1_vec + A2_vec
display(Math(rf"\mathbf{{A}} = \mathbf{{A}}' + \mathbf{{A}}'' = {latex(A_vec.evalf(5))} ~ \mathrm{{N}}"))

B_vec = B1_vec + B2_vec
display(Math(rf"\mathbf{{B}} = \mathbf{{B}}' + \mathbf{{B}}'' = {latex(B_vec.evalf(5))} ~ \mathrm{{N}}"))
```

```python
# Az eredő reakcióerők numpy tömbként
A_v = A1_v + A2_v
B_v = B1_v + B2_v

fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(16/2.54, 8/2.54))

# A = A' + A''
p = nyil(ax1, np.zeros(2), A1_v, 'darkorange', r"$\mathbf{A}'$", 1)
p = nyil(ax1, p, A2_v, 'darkorange', r"$\mathbf{A}''$", 1)
nyil(ax1, np.zeros(2), A_v, 'purple', r'$\mathbf{A}$', -1)

# B = B' + B''
p = nyil(ax2, np.zeros(2), B1_v, 'blue', r"$\mathbf{B}'$", 1)
p = nyil(ax2, p, B2_v, 'blue', r"$\mathbf{B}''$", 1)
nyil(ax2, np.zeros(2), B_v, 'purple', r'$\mathbf{B}$', -1)

for ax in (ax1, ax2):
    ax.set_aspect('equal', adjustable='datalim')
    ax.margins(0.25)
    ax.autoscale_view()
    ax.set_xlabel(r'$F_x ~ [\mathrm{N}]$')
    ax.set_ylabel(r'$F_y ~ [\mathrm{N}]$')
    ax.grid(True, alpha=0.5)
plt.tight_layout()
plt.show()
```

Látható, hogy szerkesztéssel (szuperpozícióval) ugyanazokat a reakcióerőket kaptuk, mint számítással.
