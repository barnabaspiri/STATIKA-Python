##### Feladat

A vázolt csuklós szerkezetet az $AC$ és $BC$ merev rudak alkotják, amelyek a $C$ csuklóban kapcsolódnak.

Feladatok:

a) Határozzuk meg a reakcióerőket!

b) Határozzuk meg a $BC$ rúdról a $C$ csuklóra adódó erőt!

c) Határozzuk meg az $AC$ rúdról a $C$ csuklóra adódó erőt!

##### Csomagok importálása

```python
import numpy as np # Numerikus számítások
import matplotlib.pyplot as plt # Ábrázolás
plt.rcParams['mathtext.fontset'] = 'cm' # Betűtípus az ábrákon
plt.rcParams['font.family'] = 'serif' # Betűtípus az ábrákon
plt.rcParams['font.serif'] = ['cmr10'] # Betűtípus az ábrákon
plt.rcParams['axes.formatter.use_mathtext'] = True # Betűtípus az ábrákon
plt.rcParams['axes.unicode_minus'] = False # Betűtípus az ábrákon
from IPython.display import display, Math, Latex # Szép kimenet Jupyter notebookban
import sympy as sp # Szimbólikus számítások
from sympy import latex # Szép kimenet Jupyter notebookban
sp.init_printing() # Szép kimenet Jupyter notebookban
```

##### Adatok

```python
L_1 = 4 # m
L_2 = 3 # m
L_3 = 6 # m

F_1 = 20 # kN
M_1 = 12 # kNm
```

##### Szabadtest ábra

* Mivel a $C$ csuklóban hat erő, magát a csuklót is külön szabadtest ábrával kell jellemezni! → 3 részre bontás

* A reakciókat csak 2 részre (2 rúd) bontásból is ki tudjuk számolni!

* A $C$ csuklóban ható külső erőt CSAK egy helyre vegyük fel!

<img src="img/02.jpg" alt="Feladat" width=550px>

##### A reakciók és a csuklóerők

```python
A_x, A_y, B_x, B_y = sp.symbols('A_x, A_y, B_x, B_y')
N_ACx, N_ACy, N_BCx, N_BCy = sp.symbols('N_ACx, N_ACy, N_BCx, N_BCy')
```

##### I. rész: $AC$ rúd

<img src="img/03.jpg" alt="Feladat" width=300px>

```python
# x irányú erő egyensúly
eq1 = A_x + N_ACx
display(Math(rf'\sum F_x = 0 \quad \longrightarrow \quad {latex(eq1)} = 0 \qquad (1)'))

# y irányú erő egyensúly
eq2 = A_y + N_ACy + F_1
display(Math(rf'\sum F_y = 0 \quad \longrightarrow \quad {latex(eq2)} = 0 \qquad (2)'))

# z irányú nyomatéki egyensúly a C pontra
eq3 = F_1*L_1/2 + A_y*L_1 # sp.Rational: törtként kezeli, nem float-ként
display(Math(rf'\sum M_C = 0 \quad \longrightarrow \quad {latex(eq3)} = 0 \qquad (3)'))
```

##### II. rész: $C$ csukló

<img src="img/04.jpg" alt="Feladat" width=200px>

```python
# x irányú erő egyensúly
eq4 = -N_ACx + F_1 + N_BCx
display(Math(rf'\sum F_x = 0 \quad \longrightarrow \quad {latex(eq4)} = 0 \qquad (4)'))

# y irányú erő egyensúly
eq5 = -N_ACy + N_BCy
display(Math(rf'\sum F_y = 0 \quad \longrightarrow \quad {latex(eq5)} = 0 \qquad (5)'))
```

##### III. rész: $BC$ rúd

<img src="img/05.jpg" alt="Feladat" width=220px>

```python
# x irányú erő egyensúly
eq6 = -N_BCx + B_x
display(Math(rf'\sum F_x = 0 \quad \longrightarrow \quad {latex(eq6)} = 0 \qquad (6)'))

# y irányú erő egyensúly
eq7 = -N_BCy + B_y
display(Math(rf'\sum F_y = 0 \quad \longrightarrow \quad {latex(eq7)} = 0 \qquad (7)'))

# z irányú nyomatéki egyensúly a C pontra
eq8 = M_1 + B_y*L_2 - B_x*L_3
display(Math(rf'\sum M_C = 0 \quad \longrightarrow \quad {latex(eq8)} = 0 \qquad (8)'))
```

##### Egyensúlyi egyenletek megoldása

8 egyenlet és 8 ismeretlen → megoldás

```python
sol = sp.solve([eq1, eq2, eq3, eq4, eq5, eq6, eq7, eq8], [A_x, A_y, B_x, B_y, N_ACx, N_ACy, N_BCx, N_BCy])

A_xsol = sol[A_x]
display(Math(rf'A_x = {latex(round(A_xsol, 2))} ~ \mathrm{{kN}}'))

A_ysol = sol[A_y]
display(Math(rf'A_y = {latex(round(A_ysol, 2))} ~ \mathrm{{kN}}'))

B_xsol = sol[B_x]
display(Math(rf'B_x = {latex(round(B_xsol, 2))} ~ \mathrm{{kN}}'))

B_ysol = sol[B_y]
display(Math(rf'B_y = {latex(round(B_ysol, 2))} ~ \mathrm{{kN}}'))

N_ACxsol = sol[N_ACx]
display(Math(rf'N_{{ACx}} = {latex(round(N_ACxsol, 2))} ~ \mathrm{{kN}}'))

N_ACysol = sol[N_ACy]
display(Math(rf'N_{{ACy}} = {latex(round(N_ACysol, 2))} ~ \mathrm{{kN}}'))

N_BCxsol = sol[N_BCx]
display(Math(rf'N_{{BCx}} = {latex(round(N_BCxsol, 2))} ~ \mathrm{{kN}}'))

N_BCysol = sol[N_BCy]
display(Math(rf'N_{{BCy}} = {latex(round(N_BCysol, 2))} ~ \mathrm{{kN}}'))
```

##### Ellenőrzés

A reakcióerők ismeretében az egész szerkezetre felírt egyensúlyi egyenletek ellenőrzésre használhatók.

```python
# x irányú erő egyensúly
chk1 = A_xsol + F_1 + B_xsol
display(Math(rf'\sum F_x = {latex(round(chk1, 2))}'))

# y irányú erő egyensúly
chk2 = A_ysol + F_1 + B_ysol
display(Math(rf'\sum F_y = {latex(round(chk2, 2))}'))

# z irányú nyomatéki egyensúly
chk3 = F_1*L_1*sp.Rational(1, 2) + M_1 + B_ysol*(L_1 + L_2) - B_xsol*L_3
display(Math(rf'\sum M_A = {latex(round(chk3, 2))}'))
```

Látható, hogy az egyenletek teljesülnek.

##### a) A reakcióerők

```python
A_vec = sp.Matrix([A_xsol, A_ysol])
display(Math(rf'\mathbf{{A}} = {latex(A_vec)} ~ \mathrm{{kN}}'))

B_vec = sp.Matrix([B_xsol, B_ysol])
display(Math(rf'\mathbf{{B}} = {latex(B_vec)} ~ \mathrm{{kN}}'))
```

##### b) A $BC$ rúdról a $C$ csuklóra adódó erő

```python
N_BC_vec = sp.Matrix([N_BCxsol, N_BCysol])
display(Math(rf'\mathbf{{N}}_{{BC \rightarrow C}} = {latex(sp.Matrix([N_BCx, N_BCy]))} = {latex(N_BC_vec)} ~ \mathrm{{kN}}'))
```

##### c) Az $AC$ rúdról a $C$ csuklóra adódó erő

```python
N_AC_vec = sp.Matrix([N_ACxsol, N_ACysol])
display(Math(rf'\mathbf{{N}}_{{AC \rightarrow C}} = {latex(sp.Matrix([N_ACx, N_ACy]))} = {latex(N_AC_vec)} ~ \mathrm{{kN}}'))
```

Az előjeleket a felrajzolt szabadtest ábra alapján kell értelmezni!

A II. szabadtest ábrán az $N_{BCx}$ jobbra ($+x$), az $N_{BCy}$ felfelé ($+y$), míg az $N_{ACx}$ balra ($-x$), az $N_{ACy}$ lefelé ($-y$) mutat. Így a $C$ csuklóra ható erők az $x$-$y$ koordináta-rendszerben:

```python
# A BC rúdról a C csuklóra ható erő (a II. szabadtest ábrán +x és +y irányban felvéve)
F_BC = N_BC_vec
display(Math(rf'\mathbf{{F}}_{{BC \rightarrow C}} = {latex(F_BC)} ~ \mathrm{{kN}}'))

# Az AC rúdról a C csuklóra ható erő (a II. szabadtest ábrán -x és -y irányban felvéve)
F_AC = -N_AC_vec
display(Math(rf'\mathbf{{F}}_{{AC \rightarrow C}} = {latex(F_AC)} ~ \mathrm{{kN}}'))
```
