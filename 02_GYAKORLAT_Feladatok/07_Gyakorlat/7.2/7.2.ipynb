##### Feladat

Az $AB$ és $BC$ egyenes merev rudakat a $B$ csukló kapcsolja össze. A tartó terhelése a megadott $F$ erő és $M$ nyomaték. A tartó nyugalomban van.

Feladatok:

a) Rajzolja fel a szabadtest ábrákat!

b) Határozza meg a reakcióerőket!

c) Adja meg a $BC$ rúdról az $AB$ rúdra átadódó $\mathbf{B}$ erőt!

##### Csomagok importálása

```python
import numpy as np # Numerikus számítások
import matplotlib.pyplot as plt # Ábrázolás
plt.rcParams['mathtext.fontset'] = 'cm' # Betűtípus az ábrákon
plt.rcParams['font.family'] = 'serif' # Betűtípus az ábrákon
plt.rcParams['font.serif'] = ['cmr10'] # Betűtípus az ábrákon
plt.rcParams['axes.formatter.use_mathtext'] = True # Betűtípus az ábrákon
plt.rcParams['axes.unicode_minus'] = False # Betűtípus az ábrákon
from IPython.display import display, Math, Latex # Szép kimenet Jupyter notebookban
import sympy as sp # Szimbólikus számítások
from sympy import latex # Szép kimenet Jupyter notebookban
sp.init_printing() # Szép kimenet Jupyter notebookban
```

##### Adatok

```python
L = 1 # m

F = 20 # N
M = 50 # Nm
```

##### a) Szabadtest ábrák

A szerkezetet a $B$ csuklónál 2 részre bontjuk:

##### b) A reakciók

```python
A_x, A_y, B_x, B_y, C_x, C_y = sp.symbols('A_x, A_y, B_x, B_y, C_x, C_y')
```

##### Egyensúlyi egyenletek az $AB$ rúdra

```python
# x irányú erő egyensúly
eq1 = A_x + B_x
display(Math(rf'\sum F_x = 0 \quad \longrightarrow \quad {latex(eq1)} = 0 \qquad (1)'))

# y irányú erő egyensúly
eq2 = A_y + B_y
display(Math(rf'\sum F_y = 0 \quad \longrightarrow \quad {latex(eq2)} = 0 \qquad (2)'))

# z irányú nyomatéki egyensúly a B pontra
eq3 = -A_y*2*L + A_x*2*L - M
display(Math(rf'\sum M_B = 0 \quad \longrightarrow \quad {latex(eq3)} = 0 \qquad (3)'))
```

##### Egyensúlyi egyenletek a $BC$ rúdra

```python
# x irányú erő egyensúly
eq4 = -B_x + C_x
display(Math(rf'\sum F_x = 0 \quad \longrightarrow \quad {latex(eq4)} = 0 \qquad (4)'))

# y irányú erő egyensúly
eq5 = -B_y + C_y + F
display(Math(rf'\sum F_y = 0 \quad \longrightarrow \quad {latex(eq5)} = 0 \qquad (5)'))

# z irányú nyomatéki egyensúly a B pontra
eq6 = F*L + C_y*2*L + C_x*L
display(Math(rf'\sum M_B = 0 \quad \longrightarrow \quad {latex(eq6)} = 0 \qquad (6)'))
```

##### Egyensúlyi egyenletek megoldása

Az (1) és (2) egyenletet beírva a (3)-ba, illetve a (4) és (5) egyenletet beírva a (6)-ba 2 ismeretlenes ($B_x$, $B_y$) egyenletrendszert kapunk:

```python
# (1) és (2) beírva (3)-ba
eq7 = eq3.subs({A_x: sp.solve(eq1, A_x)[0], A_y: sp.solve(eq2, A_y)[0]})
display(Math(rf'{latex(eq7)} = 0 \qquad (7)'))

# (4) és (5) beírva (6)-ba
eq8 = eq6.subs({C_x: sp.solve(eq4, C_x)[0], C_y: sp.solve(eq5, C_y)[0]})
display(Math(rf'{latex(eq8)} = 0 \qquad (8)'))
```

```python
sol = sp.solve([eq7, eq8], [B_x, B_y])

B_xsol = sol[B_x]
display(Math(rf'B_x = {latex(round(B_xsol, 2))} ~ \mathrm{{N}}'))

B_ysol = sol[B_y]
display(Math(rf'B_y = {latex(round(B_ysol, 2))} ~ \mathrm{{N}}'))
```

Így a többi reakció:

```python
# (1)-ből
A_xsol = sp.solve(eq1.subs(B_x, B_xsol), A_x)[0]
display(Math(rf'A_x = {latex(round(A_xsol, 2))} ~ \mathrm{{N}}'))

# (2)-ből
A_ysol = sp.solve(eq2.subs(B_y, B_ysol), A_y)[0]
display(Math(rf'A_y = {latex(round(A_ysol, 2))} ~ \mathrm{{N}}'))

# (4)-ből
C_xsol = sp.solve(eq4.subs(B_x, B_xsol), C_x)[0]
display(Math(rf'C_x = {latex(round(C_xsol, 2))} ~ \mathrm{{N}}'))

# (5)-ből
C_ysol = sp.solve(eq5.subs(B_y, B_ysol), C_y)[0]
display(Math(rf'C_y = {latex(round(C_ysol, 2))} ~ \mathrm{{N}}'))
```

##### A reakcióerők

```python
A_vec = sp.Matrix([A_xsol, A_ysol])
display(Math(rf'\mathbf{{A}} = {latex(A_vec)} ~ \mathrm{{N}}'))

C_vec = sp.Matrix([C_xsol, C_ysol])
display(Math(rf'\mathbf{{C}} = {latex(C_vec)} ~ \mathrm{{N}}'))
```

##### c) A $BC$ rúdról az $AB$ rúdra átadódó erő

```python
B_vec = sp.Matrix([B_xsol, B_ysol])
display(Math(rf'\mathbf{{B}} = {latex(B_vec)} ~ \mathrm{{N}}'))
```

Ez a $\mathbf{B}$ erő adódik át a $BC$ rúdról az $AB$ rúdra!

##### Ellenőrzés

A reakcióerők ismeretében az egész szerkezetre felírt egyensúlyi egyenletek ellenőrzésre használhatók.

```python
# x irányú erő egyensúly
chk1 = A_xsol + C_xsol
display(Math(rf'\sum F_x = {latex(round(chk1, 2))}'))

# y irányú erő egyensúly
chk2 = A_ysol + F + C_ysol
display(Math(rf'\sum F_y = {latex(round(chk2, 2))}'))

# z irányú nyomatéki egyensúly
chk3 = -M + F*3*L + C_ysol*4*L - C_xsol*L
display(Math(rf'\sum M_A = {latex(round(chk3, 2))}'))
```
