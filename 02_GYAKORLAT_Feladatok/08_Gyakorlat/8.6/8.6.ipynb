##### Csomagok importálása

```python
import numpy as np # Numerikus számítások
import matplotlib.pyplot as plt # Ábrázolás
plt.rcParams['mathtext.fontset'] = 'cm' # Betűtípus az ábrákon
plt.rcParams['font.family'] = 'serif' # Betűtípus az ábrákon
plt.rcParams['font.serif'] = ['cmr10'] # Betűtípus az ábrákon
plt.rcParams['axes.formatter.use_mathtext'] = True # Betűtípus az ábrákon
plt.rcParams['axes.unicode_minus'] = False # Betűtípus az ábrákon
from IPython.display import display, Math, Latex # Szép kimenet Jupyter notebookban
import sympy as sp # Szimbólikus számítások
from sympy import latex # Szép kimenet Jupyter notebookban
sp.init_printing() # Szép kimenet Jupyter notebookban
```

##### Adatok

```python
L = 4 # m

p = 3 # kN/m
F_1 = 3 # kN

# Rúdirányú koordináta
x = sp.symbols('x')
```

##### Előjelkonvenció

Az igénybevételek előjelét az alábbi előjelkonvenció szerint határozzuk meg (lásd a 8.1 feladatot):

<img src="img/02.jpg" alt="Feladat" width=300px>

##### Redukálás a $K$ keresztmetszetbe

* Jobbról redukálunk → nem kell a reakciókat meghatározni, mert a jobb oldalon a teljes erőrendszer ismert.

* A $K$ keresztmetszettől jobbra lévő részen az $F_1$ erő és a megoszló terhelés hat. A megoszló terhelést az eredőjével ($F_p$) helyettesítjük, amely a $K$ keresztmetszettől $(L - x)/2$ távolságra hat.

<img src="img/03.jpg" alt="Feladat" width=400px>

##### Az igénybevételi függvények

```python
# Normálerő: nincs rúdirányú erő
N = sp.Integer(0)
display(Math(rf'N(x) = {latex(N)}'))

# Nyíróerő: az F_1 erő felfelé mutat (jobbról nézve negatív), az F_p erő lefelé mutat (jobbról nézve pozitív)
V = -F_1 + p*(L - x)
display(Math(rf'V(x) = -F_1 + p (L - x) = {latex(V)} ~ \mathrm{{kN}} \qquad \text{{ha }} x ~ [\mathrm{{m}}]'))

# Hajlítónyomaték: az F_1 nyomatéka a K pontra az óramutató járásával ellentétes (jobbról nézve negatív),
# az F_p nyomatéka az óramutató járásával megegyező (jobbról nézve pozitív)
M_h = sp.expand(-F_1*(L - x) + p*(L - x)**2/2)
display(Math(rf'M_h(x) = -F_1 (L - x) + p \frac{{(L - x)^2}}{{2}} = {latex(M_h)} ~ \mathrm{{kNm}} \qquad \text{{ha }} x ~ [\mathrm{{m}}]'))

# Csavarónyomaték: nincs rúdirányú nyomaték
M_t = sp.Integer(0)
display(Math(rf'M_t(x) = {latex(M_t)}'))
```

##### A reakciók

Ha kiszámoljuk a reakciókat, az igénybevételi függvényeket a befogásnál ellenőrizni tudjuk. A befogásban ható reakciók: $A_x$ ($+x$ irányban), $A_y$ ($+y$ irányban) és $M_A$ (az óramutató járásával ellentétes irányban felvéve).

```python
A_x, A_y, M_A = sp.symbols('A_x, A_y, M_A')
```

##### Egyensúlyi egyenletek

```python
# x irányú erő egyensúly
eq1 = A_x
display(Math(rf'\sum F_x = 0 \quad \longrightarrow \quad {latex(eq1)} = 0 \qquad (1)'))

# y irányú erő egyensúly
eq2 = A_y - p*L + F_1
display(Math(rf'\sum F_y = 0 \quad \longrightarrow \quad {latex(eq2)} = 0 \qquad (2)'))

# z irányú nyomatéki egyensúly az A pontra
eq3 = M_A - p*L*L*sp.Rational(1, 2) + F_1*L
display(Math(rf'\sum M_A = 0 \quad \longrightarrow \quad {latex(eq3)} = 0 \qquad (3)'))
```

##### Egyensúlyi egyenletek megoldása

```python
sol = sp.solve([eq1, eq2, eq3], [A_x, A_y, M_A])

A_xsol = sol[A_x]
display(Math(rf'A_x = {latex(A_xsol)} ~ \mathrm{{kN}}'))

A_ysol = sol[A_y]
display(Math(rf'A_y = {latex(A_ysol)} ~ \mathrm{{kN}}'))

M_Asol = sol[M_A]
display(Math(rf'M_A = {latex(M_Asol)} ~ \mathrm{{kNm}}'))
```

##### Igénybevételek a befogásnál ($x = 0$)

A befogás keresztmetszetére balról a reakciók, jobbról az igénybevételek hatnak:

<img src="img/04.jpg" alt="Feladat" width=300px>

```python
V_0 = V.subs(x, 0)
display(Math(rf'V(x=0) = {latex(V_0)} ~ \mathrm{{kN}}'))

M_h0 = M_h.subs(x, 0)
display(Math(rf'M_h(x=0) = {latex(M_h0)} ~ \mathrm{{kNm}}'))

# A befogás keresztmetszetének egyensúlya
display(Math(rf'A_y - V(x=0) = {latex(A_ysol - V_0)}'))
display(Math(rf'M_A - M_h(x=0) = {latex(M_Asol - M_h0)}'))
```

Látható, hogy teljesül az egyensúly.

##### Derivált kapcsolatok

$V'(x) = p(x)$ és $M_h'(x) = -V(x)$.

A megoszló terhelés lefelé ($-y$ irányba) mutat, így $p(x) = -p$.

```python
# A megoszló terhelés függvénye
p_x = -p
display(Math(rf'p(x) = {latex(p_x)} ~ \mathrm{{kN/m}}'))

# A nyíróerő deriváltja
dV = sp.diff(V, x)
display(Math(rf"V'(x) = {latex(dV)} \qquad p(x) = {latex(p_x)}"))

# A hajlítónyomaték deriváltja
dM_h = sp.diff(M_h, x)
display(Math(rf"M_h'(x) = {latex(dM_h)} \qquad -V(x) = {latex(-V)}"))
```

Látható, hogy az összefüggések teljesülnek.

##### Ábrázolás

* $V(x)$: könnyű, lineáris egyenes, 2 pontot kiszámolunk, behúzzuk.

* $M_h(x)$: parabola, szerkeszteni kell.

Python segítségével a parabola közvetlenül is ábrázolható, de kézi rajzolásnál szerkeszteni kell. A szerkesztés lépései:

1. $p(x)$ és $V(x)$ felrajzolása, segédvonal a lineáris $V(x)$ tartomány felénél (ez most $x = 2 ~ \mathrm{m}$)
2. $M_h(x)$ kezdő- és végpontbeli értékének felrajzolása
3. $M_h(x)$ kezdő- és végpontbeli érintőinek berajzolása
4. érintők metszéspontjának kiszámítása: $P$ pont
5. húzzuk be a húrt $M_h(x=0)$ és $M_h(x=L)$ közé → $Q$ pont: a húr szakasz felénél lévő pontja
6. felezzük meg a $PQ$ szakaszt: a szakasz felezőpontja az $S$ pont
7. húzzunk párhuzamost a húrral, $S$ ponton keresztül → ez a parabola érintője $S$ pontban
8. rajzoljuk meg a parabolát!

```python
# 1.) Segédvonal a lineáris V(x) tartomány felénél
x_F = sp.Rational(L, 2)
display(Math(rf'x_F = \frac{{L}}{{2}} = {latex(x_F)} ~ \mathrm{{m}}'))

# 2.) M_h(x) kezdő- és végpontbeli értéke
M_h0 = M_h.subs(x, 0)
display(Math(rf'M_h(x=0) = {latex(M_h0)} ~ \mathrm{{kNm}}'))

M_hL = M_h.subs(x, L)
display(Math(rf'M_h(x=L) = {latex(M_hL)} ~ \mathrm{{kNm}}'))

# 3.) M_h(x) kezdő- és végpontbeli érintőinek meredeksége: M_h'(x) = -V(x)
m_0 = -V.subs(x, 0)
display(Math(rf"M_h'(x=0) = -V(x=0) = {latex(m_0)}"))

m_L = -V.subs(x, L)
display(Math(rf"M_h'(x=L) = -V(x=L) = {latex(m_L)}"))
```

4.) Az érintők metszéspontja ($P$ pont):

* $A_1$ és $A_2$ téglalapok berajzolása (a $V(x)$ ábrán a tartomány két felén).

* A kezdőpontbeli érintő az $L/2$ hosszú szakaszon $M_h'(x=0) \cdot L/2 = -V(x=0) \cdot L/2 = -A_1$ értékkel változik → a $P$ pont helye: $M_h(x)$ kezdőértékéből ($M_h(x=0)$) vonjuk le $A_1$-et!

* Ugyanígy a végpontból: $M_h(x=L) + A_2$.

```python
# 4.) Az A_1 és A_2 téglalapok
A_1 = V.subs(x, 0)*x_F
display(Math(rf'A_1 = V(x=0) \cdot \frac{{L}}{{2}} = {latex(A_1)} ~ \mathrm{{kNm}}'))

A_2 = V.subs(x, L)*x_F
display(Math(rf'A_2 = V(x=L) \cdot \frac{{L}}{{2}} = {latex(A_2)} ~ \mathrm{{kNm}}'))

# 4.) Az érintők metszéspontja: P pont
r_P = sp.Matrix([x_F, M_h0 - A_1])
display(Math(rf'P = {latex(r_P)}'))

# Ellenőrzés: a végpontból is ugyanazt kapjuk
display(Math(rf'M_h(x=L) + A_2 = {latex(M_hL + A_2)} ~ \mathrm{{kNm}}'))
```

```python
# 5.) A húr M_h(x=0) és M_h(x=L) között, Q pont: a húr szakasz felénél lévő pontja
r_Q = sp.Matrix([x_F, (M_h0 + M_hL)/2])
display(Math(rf'Q = {latex(r_Q)}'))

# 6.) A PQ szakasz felezőpontja: S pont
r_S = (r_P + r_Q)/2
display(Math(rf'S = {latex(r_S)}'))

# 7.) A húrral párhuzamos egyenes az S ponton keresztül: a parabola érintője S pontban
m_hur = (M_hL - M_h0)/L
display(Math(rf'm_{{\text{{húr}}}} = {latex(m_hur)}'))
```

Ellenőrzés: az $S$ pont rajta van a parabolán, és ott az érintő meredeksége megegyezik a húr meredekségével:

```python
display(Math(rf'M_h(x={latex(x_F)}) = {latex(M_h.subs(x, x_F))} ~ \mathrm{{kNm}}'))
display(Math(rf"M_h'(x={latex(x_F)}) = {latex(dM_h.subs(x, x_F))}"))
```

8.) Rajzoljuk meg a parabolát!

* ahol $V(x) = 0$, ott $M_h(x)$-nek szélsőértéke van (vízszintes érintő)

* adott érintők: kezdőpont, végpont, $S$ pont

```python
# 8.) A szélsőérték helye: V(x) = 0
x_sz = sp.solve(V, x)[0]
display(Math(rf'V(x) = 0 \quad \longrightarrow \quad x = {latex(x_sz)} ~ \mathrm{{m}}'))

M_hsz = M_h.subs(x, x_sz)
display(Math(rf'M_h(x={latex(x_sz)}) = {latex(round(M_hsz, 2))} ~ \mathrm{{kNm}}'))
```

A szerkesztés léptékhelyesen:

```python
def igenybeveteli_abra(ax, f, x_0, x_1, felirat, szin='red'):
    # A függvény kiértékelése numerikusan az [x_0, x_1] tartományon
    x_vals = np.linspace(x_0, x_1, 201)
    f_vals = sp.lambdify(x, f, 'numpy')(x_vals) * np.ones_like(x_vals) # Konstans függvény esetén is tömböt kapunk
    # A függvény görbéje (zorder: az x tengely fölé rajzoljuk)
    ax.plot(x_vals, f_vals, color=szin, lw=1.5, zorder=3)
    # Satírozás a görbe és az x tengely között (mint a kézi rajzokon)
    ax.vlines(x_vals[::10], 0, f_vals[::10], color=szin, lw=0.8)
    # Lezárás a tartomány két végén
    ax.vlines([x_0, x_1], 0, [f_vals[0], f_vals[-1]], color=szin, lw=1.5, zorder=3)
    # Az x tengely
    ax.axhline(0, color='black', lw=1)
    ax.set_ylabel(felirat)
    ax.grid(True, alpha=0.5)
```

```python
# A szerkesztés pontjai numpy tömbként
P_v = np.array(r_P, dtype=float).flatten()
Q_v = np.array(r_Q, dtype=float).flatten()
S_v = np.array(r_S, dtype=float).flatten()

fig, axs = plt.subplots(3, 1, figsize=(14/2.54, 20/2.54), sharex=True, gridspec_kw={'height_ratios': [1, 2, 3]})

# 1.) p(x)
igenybeveteli_abra(axs[0], p_x, 0, L, r'$p(x) ~ [\mathrm{kN/m}]$')
axs[0].set_ylim(-4.5, 1.5)

# 1.) V(x) és 4.) az A_1, A_2 téglalapok
igenybeveteli_abra(axs[1], V, 0, L, r'$V(x) ~ [\mathrm{kN}]$')
axs[1].add_patch(plt.Rectangle((0, 0), float(x_F), float(V.subs(x, 0)), fill=False, ls='--', color='purple'))
axs[1].add_patch(plt.Rectangle((float(x_F), 0), float(L - x_F), float(V.subs(x, L)), fill=False, ls='--', color='purple'))
axs[1].annotate(r'$A_1$', xy=(float(x_F)/2, float(V.subs(x, 0))), xytext=(0, 5), textcoords='offset points', ha='center', color='purple', fontsize=12)
axs[1].annotate(r'$A_2$', xy=((float(x_F) + L)/2, float(V.subs(x, L))), xytext=(0, -16), textcoords='offset points', ha='center', color='purple', fontsize=12)
axs[1].plot(float(x_sz), 0, 'o', color='green', zorder=3)
axs[1].annotate(rf'$x = {latex(x_sz)}$', xy=(float(x_sz), 0), xytext=(4, 6), textcoords='offset points', color='green', fontsize=12)
axs[1].set_ylim(-6, 12)

# 2.) M_h(x)
igenybeveteli_abra(axs[2], M_h, 0, L, r'$M_h(x) ~ [\mathrm{kNm}]$')

# 3.) Érintők a kezdő- és végpontban, 4.) metszéspontjuk a P pont
axs[2].plot([0, P_v[0], L], [float(M_h0), P_v[1], float(M_hL)], color='darkorange', lw=1.2)
axs[2].annotate(rf'$m = {latex(m_0)}$', xy=(1.5, float(M_h0 + m_0*1.5)), xytext=(-50, -22), textcoords='offset points', color='darkorange', fontsize=12)
axs[2].annotate(rf'$m = {latex(m_L)}$', xy=(3, float(M_hL + m_L*(3 - L))), xytext=(-40, -32), textcoords='offset points', color='darkorange', fontsize=12)

# 5.) Húr
axs[2].plot([0, L], [float(M_h0), float(M_hL)], color='green', lw=1.2)
axs[2].annotate(r'$\mathrm{h\acute{u}r}$', xy=(1, float(M_h0 + m_hur*1)), xytext=(4, 6), textcoords='offset points', color='green', fontsize=12)

# 7.) Érintő az S pontban (párhuzamos a húrral)
x_e = np.array([S_v[0] - 1.3, S_v[0] + 1.3])
axs[2].plot(x_e, S_v[1] + float(m_hur)*(x_e - S_v[0]), color='green', ls='--', lw=1.2)

# 4.), 5.), 6.) P, Q és S pontok (zorder: minden más fölé rajzoljuk)
for nev, r_v in zip(['P', 'Q', 'S'], [P_v, Q_v, S_v]):
    axs[2].plot(r_v[0], r_v[1], 'o', color='blue', zorder=10)
    axs[2].annotate(rf'${nev}$', xy=r_v, xytext=(6, 4), textcoords='offset points', color='blue', fontsize=12, zorder=10)
axs[2].set_ylim(-8, 14)

# Segédvonalak: a tartomány fele és a szélsőérték helye
for ax in axs[1:]:
    ax.axvline(float(x_F), color='gray', ls='--', lw=0.8)
    ax.axvline(float(x_sz), color='gray', ls=':', lw=0.8)

# A függvények ábrázolása 0-tól L-ig
axs[2].set_xlim(-0.2, L+0.2)
axs[2].set_xlabel(r'$x ~ [\mathrm{m}]$')
plt.tight_layout()
plt.savefig('igenybeveteli_fuggvenyek.pdf') # Az ábra mentése PDF dokumentumba
plt.show()
```
