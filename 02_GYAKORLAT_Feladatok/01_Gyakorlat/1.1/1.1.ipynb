##### Feladat (Példatár 1.1)

Egy merev testet az ábrán megadott $F_1$ és $F_2$ erők terhelik. Helyettesítsük a két erőt az eredőjével, és adjuk meg az eredő erő hatásvonalának egy pontját! Milyen szöget zár be az eredő hatásvonala az $x$-tengellyel?

##### Csomagok importálása

```python
import numpy as np # Numerikus számítások
import matplotlib.pyplot as plt # Ábrázolás
plt.rcParams['mathtext.fontset'] = 'cm' # Betűtípus az ábrákon
plt.rcParams['font.family'] = 'serif' # Betűtípus az ábrákon
plt.rcParams['font.serif'] = ['cmr10'] # Betűtípus az ábrákon
plt.rcParams['axes.formatter.use_mathtext'] = True # Betűtípus az ábrákon
plt.rcParams['axes.unicode_minus'] = False # Betűtípus az ábrákon
from IPython.display import display, Math, Latex # Szép kimenet Jupyter notebookban
import sympy as sp # Szimbólikus számítások
from sympy import latex # Szép kimenet Jupyter notebookban
sp.init_printing() # Szép kimenet Jupyter notebookban
```

##### Erővektorok és azok támadáspontjuk

```python
# Első erő
r_1 = sp.Matrix([1, 1]) # m
F_1 = sp.Matrix([60, 20]) # N

# Második erő
r_2 = sp.Matrix([8, 5.5]) # m
F_2 = sp.Matrix([-20, 15]) # N
```

##### Az eredő erő számítása

```python
# Vektorok összege
F_e = F_1 + F_2
display(Math(rf'\mathbf{{F}}_e = {latex(F_e)}'))
```

##### Az eredő erő támadáspontjának számítása

```python
# Ismeretlenek
lam1, lam2 = sp.symbols('λ_1, λ_2')

# Eredő támadáspontja a két vektor segítségével
r_c1 = r_1 + lam1 * F_1
r_c2 = r_2 + lam2 * F_2

# Két egyenlet, két ismeretlen
print('Az eredő támadáspontja:')
display(Math(rf'\mathbf{{r}}_c = {latex(r_c1)} = {latex(r_c2)}'))

# Két egyenlet definiálása
eq1 = r_c1[0] - r_c2[0]
eq2 = r_c1[1] - r_c2[1]

# Megoldás két ismeretlenre
sol = sp.solve([eq1, eq2], [lam1, lam2])

lam1_sol = sol[lam1]
lam2_sol = sol[lam2]

print('\nMegoldva két a ismeretlenre:')
display(Math(rf'\lambda_1 = {latex(lam1_sol)} ~ \mathrm{{m/N}}'))
display(Math(rf'\lambda_2 = {latex(lam2_sol)} ~ \mathrm{{m/N}}'))
```

##### Az eredő hatásvonala és az $x$ tengely által bezárt szög

```python
# A bezárt szög 
alpha = sp.atan(F_e[1] / F_e[0]) # rad
alpha = alpha * 180 / np.pi

display(Math(rf'\alpha = {latex(alpha.evalf(4))}^\circ'))
```
