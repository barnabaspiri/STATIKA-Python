##### Feladat (Példatár 1.6)

A $G = 500$ N súlyú rúd $A$ vége csuklósan rögzített, $B$ vége a teljesen sima falnak támaszkodik. Határozzuk meg a reakcióerőket szerkesztéssel és számítással!

##### Csomagok importálása

```python
import numpy as np # Numerikus számítások
import matplotlib.pyplot as plt # Ábrázolás
plt.rcParams['mathtext.fontset'] = 'cm' # Betűtípus az ábrákon
plt.rcParams['font.family'] = 'serif' # Betűtípus az ábrákon
plt.rcParams['font.serif'] = ['cmr10'] # Betűtípus az ábrákon
plt.rcParams['axes.formatter.use_mathtext'] = True # Betűtípus az ábrákon
plt.rcParams['axes.unicode_minus'] = False # Betűtípus az ábrákon
from IPython.display import display, Math, Latex # Szép kimenet Jupyter notebookban
import sympy as sp # Szimbólikus számítások
from sympy import latex # Szép kimenet Jupyter notebookban
sp.init_printing() # Szép kimenet Jupyter notebookban
```

##### Adatok

```python
L_1 = 1 # m
L_2 = 4 # m
G = 500 # N
```

##### Az erővektorok felírása

```python
# A rúd vízszintessel bezárt szöge a geometria alapján
alpha = sp.atan(L_2 / L_1)
display(Math(rf'\alpha = {latex(sp.deg(alpha).evalf(4))}^\circ'))
```

```python
F_A, F_B = sp.symbols('F_A, F_B')

F_Avec = sp.Matrix([F_A*sp.cos(alpha), F_A*sp.sin(alpha)])
display(Math(rf'\mathbf{{F}}_A = {latex(F_Avec.evalf(4))}'))

G_vec = sp.Matrix([0, -G])
display(Math(rf'\mathbf{{G}} = {latex(G_vec.evalf(4))}'))

F_Bvec = sp.Matrix([-F_B, 0])
display(Math(rf'\mathbf{{F}}_B = {latex(F_Bvec)}'))
```

##### Az eredő erő számítása

```python
F_e = F_Avec + G_vec + F_Bvec
display(Math(rf'\mathbf{{F}}_e = {latex(F_e.evalf(4))}'))
```

```python
# Az y komponensből számolható F_A
F_Asol = sp.solve(F_e[1], F_A)[0]
display(Math(rf'F_A= {latex(F_Asol.evalf(5))} ~ N'))
```

```python
# Az x komponensből számolható F_B
F_Bsol = sp.solve(F_e[0].subs(F_A, F_Asol), F_B)[0]
display(Math(rf'F_B= {latex(F_Bsol.evalf(5))} ~ N'))
```

##### Tehát a vektorok

```python
F_A, F_B = sp.symbols('F_A, F_B')

# F_A
F_Avec = F_Avec.subs(F_A, F_Asol).subs(F_B, F_Bsol)
display(Math(rf'\mathbf{{F}}_A = {latex(F_Avec.evalf(6))}  ~ N'))

# G
display(Math(rf'\mathbf{{G}} = {latex(G_vec.evalf(6))}  ~ N'))

# F_B
F_Bvec = F_Bvec.subs(F_A, F_Asol).subs(F_B, F_Bsol)
display(Math(rf'\mathbf{{F}}_B = {latex(F_Bvec.evalf(6))} ~ N'))
```
