##### Feladat (Példatár 1.3)

Mekkora legyen az $F_1$ erő nagysága ha azt szeretnénk, hogy a kötelekről átadódó erők eredőjének ne legyen függőleges komponense? Mekkora ekkor a kötelekről átadódó eredő erő $x$-irányú összetevője? Az $F_2$ erő nagysága 2 kN.

##### Csomagok importálása

```python
import numpy as np # Numerikus számítások
import matplotlib.pyplot as plt # Ábrázolás
plt.rcParams['mathtext.fontset'] = 'cm' # Betűtípus az ábrákon
plt.rcParams['font.family'] = 'serif' # Betűtípus az ábrákon
plt.rcParams['font.serif'] = ['cmr10'] # Betűtípus az ábrákon
plt.rcParams['axes.formatter.use_mathtext'] = True # Betűtípus az ábrákon
plt.rcParams['axes.unicode_minus'] = False # Betűtípus az ábrákon
from IPython.display import display, Math, Latex # Szép kimenet Jupyter notebookban
import sympy as sp # Szimbólikus számítások
from sympy import latex # Szép kimenet Jupyter notebookban
sp.init_printing() # Szép kimenet Jupyter notebookban
```

##### Adatok

```python
F_2 = 2000 # N
alpha = -30 # deg

alpha = sp.rad(alpha) # rad
```

##### Az erővektorok

```python
F_1 = sp.Symbol('F_1')

F_1vec = sp.Matrix([F_1*sp.cos(alpha), F_1*sp.sin(alpha)])
display(Math(rf'\mathbf{{F}}_1 = {latex(F_1vec)}'))

F_2vec = sp.Matrix([0, F_2])
display(Math(rf'\mathbf{{F}}_2 = {latex(F_2vec)}'))
```

##### Az eredő erő számítása

```python
F_e = F_1vec + F_2vec
display(Math(rf'\mathbf{{F}}_e = {latex(F_e)}'))
```

##### Ne legyen függőleges komponens

```python
# F_ey = 0
eq = F_e[1]

sol = sp.solve(eq, F_1)
F_1sol = sol[0]

display(Math(rf'F_1 = {latex(F_1sol)} ~ N'))
```

##### Így az eredő

```python
F_e = F_e.subs(F_1, F_1sol)
display(Math(rf'\mathbf{{F}}_e = {latex(F_e.evalf(5))} ~ N'))
```
