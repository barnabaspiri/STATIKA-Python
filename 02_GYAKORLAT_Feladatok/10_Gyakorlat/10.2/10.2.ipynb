##### Csomagok importálása

```python
import numpy as np # Numerikus számítások
import matplotlib.pyplot as plt # Ábrázolás
plt.rcParams['mathtext.fontset'] = 'cm' # Betűtípus az ábrákon
plt.rcParams['font.family'] = 'serif' # Betűtípus az ábrákon
plt.rcParams['font.serif'] = ['cmr10'] # Betűtípus az ábrákon
plt.rcParams['axes.formatter.use_mathtext'] = True # Betűtípus az ábrákon
plt.rcParams['axes.unicode_minus'] = False # Betűtípus az ábrákon
from IPython.display import display, Math, Latex # Szép kimenet Jupyter notebookban
import sympy as sp # Szimbólikus számítások
from sympy import latex # Szép kimenet Jupyter notebookban
sp.init_printing() # Szép kimenet Jupyter notebookban
```

##### Adatok

```python
# A feladat paraméteresen adott, így az adatokat szimbólumként vesszük fel
F, R = sp.symbols('F, R', positive=True)

# Leíró koordináták: a görbe rúdon a φ szög, az egyenes rúdon az y koordináta
phi, y = sp.symbols('varphi, y')
```

##### Egyensúly

A rúdszerkezetnek nincs megtámasztása, csak a két $F$ erő terheli: a bal oldali szabad végen ($A$) $-x$ irányú, a kör középpontjában ($O$, az egyenes rúd végén) $+x$ irányú $F$ erő hat. Ellenőrizzük, hogy a rúdszerkezet egyensúlyban van-e:

```python
# Az erők támadáspontjainak helyvektorai (az origó a kör középpontjában van)
r_A = sp.Matrix([-R, 0, 0])
r_O = sp.Matrix([0, 0, 0])

# Az erővektorok
F_A = sp.Matrix([-F, 0, 0])
F_O = sp.Matrix([F, 0, 0])

# Erő egyensúly
display(Math(rf'\sum \mathbf{{F}} = \mathbf{{F}}_A + \mathbf{{F}}_O = {latex(F_A + F_O)}'))

# Nyomatéki egyensúly az origóra
display(Math(rf'\sum \mathbf{{M}}_O = \mathbf{{r}}_A \times \mathbf{{F}}_A + \mathbf{{r}}_O \times \mathbf{{F}}_O = {latex(r_A.cross(F_A) + r_O.cross(F_O))}'))
```

A két erő egy hatásvonalon (az $x$ tengelyen) hat, egyenlő nagyságú és ellentétes irányú → a rúdszerkezet egyensúlyban van, reakciók nem ébrednek.

##### Szakaszok

* A görbe rúd és az egyenes rúd csatlakozásánál (a kör alsó pontjában) a rúd iránya megtörik → 2 szakaszra kell bontani az igénybevételi függvényeket!

* I. szakasz (görbe rúd): $\varphi \in [0^\circ, 270^\circ]$, azaz $\varphi \in [0, \frac{3}{2}\pi]$

* II. szakasz (egyenes rúd): $y \in [0, R]$

##### Előjelkonvenció

* A rudat a bal oldali szabad végtől kezdve járjuk körbe: a görbe rúdon a $\varphi$ szög az óramutató járásával megegyező irányban nő, az egyenes rúdon alulról felfelé haladunk ($y$ koordináta).

* Mindkét szakaszon balról redukálunk (a körbejárás szerint a keresztmetszet előtti részt), így a keresztmetszettől balra csak a bal oldali $F$ erő hat.

* Az előjelkonvenció a görbe rúdon a görbületi középpont felől nézve értendő, és fordul a körbejárással (lásd a 10.1 feladatot).

##### Az igénybevételi függvények

##### I. szakasz: $\varphi \in [0, 3\pi/2]$ (görbe rúd)

A $K$ keresztmetszettől balra (a körbejárás szerint előtte) csak a bal oldali $F$ erő hat. Ezt redukáljuk a $K$ keresztmetszet súlypontjába, és normál és sugár irányú komponensekre bontjuk:

<img src="img/03.jpg" alt="Feladat" width=300px>

```python
# Az F erő komponensei: sugár irányú (nyíróerő) és normál irányú (normálerő)
F_V = F*sp.cos(phi)
display(Math(rf'F_V = {latex(F_V)}'))

F_N = F*sp.sin(phi)
display(Math(rf'F_N = {latex(F_N)}'))

# Az F erő erőkarja: a K keresztmetszet (előjeles) távolsága az x tengelytől, az F hatásvonalától
k = R*sp.sin(phi)
display(Math(rf'k = {latex(k)}'))
```

Balról redukálunk: a normálerő pozitív, ha húzza a keresztmetszetet, a nyíróerő pozitív, ha a görbületi középponttól elfelé mutat, a hajlítónyomaték pedig pozitív, ha az óramutató járásával ellentétes.

```python
# Normálerő: az F_N komponens (pozitív értéknél) húzza a keresztmetszetet, ami balról nézve pozitív
N_1 = F_N
display(Math(rf'N_1(\varphi) = F_N = {latex(N_1)}'))

# Nyíróerő: az F_V komponens (pozitív értéknél) a görbületi középponttól elfelé mutat, ami balról nézve pozitív
V_1 = F_V
display(Math(rf'V_1(\varphi) = F_V = {latex(V_1)}'))

# Hajlítónyomaték: az F erő nyomatéka a K pontra (pozitív erőkarnál) az óramutató járásával megegyező, ami balról nézve negatív
M_h1 = -F*k
display(Math(rf'M_{{h1}}(\varphi) = -F \cdot k = {latex(M_h1)}'))

# Csavarónyomaték: nincs rúdirányú nyomaték
M_t1 = sp.Integer(0)
display(Math(rf'M_{{t1}}(\varphi) = {latex(M_t1)}'))
```

##### II. szakasz: $y \in [0, R]$ (egyenes rúd)

A $K$ keresztmetszettől balra (a körbejárás szerint előtte) a teljes görbe rúd van, amelyen csak a bal oldali $F$ erő hat. Az $F$ erő merőleges az egyenes rúdra, erőkarja a $K$ keresztmetszet és az $x$ tengely távolsága: $R - y$.

```python
# Normálerő: az F erő merőleges a rúdra, nincs rúdirányú komponense
N_2 = sp.Integer(0)
display(Math(rf'N_2(y) = {latex(N_2)}'))

# Nyíróerő: az F erő -x irányba mutat, ami az előjelkonvenció szerint pozitív
V_2 = F
display(Math(rf'V_2(y) = {latex(V_2)}'))

# Hajlítónyomaték: az F erő nyomatéka a K pontra (erőkar: R - y) az óramutató járásával ellentétes, ami az előjelkonvenció szerint pozitív
M_h2 = F*(R - y)
display(Math(rf'M_{{h2}}(y) = F \cdot (R - y) = {latex(M_h2)}'))

# Csavarónyomaték: nincs rúdirányú nyomaték
M_t2 = sp.Integer(0)
display(Math(rf'M_{{t2}}(y) = {latex(M_t2)}'))
```

##### Ellenőrzés

* A sarokpontban ($\varphi = 3\pi/2$, ill. $y = 0$) a rúd iránya $90^\circ$-ot fordul: a görbe rúd végén az érintő vízszintes, az egyenes rúd függőleges → a normálerő és a nyíróerő szerepet cserél: $V_2(0) = -N_1(3\pi/2)$ és $N_2(0) = V_1(3\pi/2)$. A hajlítónyomaték folytonos, mert ott nem hat koncentrált nyomaték.

* Az egyenes rúd végén ($y = R$) jobbról csak a kör középpontjában ható $F$ erő hat, amely merőleges a rúdra, és a keresztmetszetben támad: $N = 0$, $V = F$ és $M_h = 0$.

* Derivált kapcsolatok: a görbe rúdon a $\varphi$ az óramutató járásával megegyezik, így $\dfrac{\mathrm{d} M_h(\varphi)}{\mathrm{d}(R \varphi)} = -V(\varphi)$, az egyenes rúdon pedig $M_h'(y) = -V(y)$.

```python
# A sarokpontban: a normálerő és a nyíróerő szerepet cserél
display(Math(rf'N_1(\varphi=3\pi/2) = {latex(N_1.subs(phi, 3*sp.pi/2))} \qquad V_2(y=0) = {latex(V_2)}'))
display(Math(rf'V_1(\varphi=3\pi/2) = {latex(V_1.subs(phi, 3*sp.pi/2))} \qquad N_2(y=0) = {latex(N_2)}'))

# A sarokpontban: a hajlítónyomaték folytonos
display(Math(rf'M_{{h1}}(\varphi=3\pi/2) = {latex(M_h1.subs(phi, 3*sp.pi/2))} \qquad M_{{h2}}(y=0) = {latex(M_h2.subs(y, 0))}'))

# Az igénybevételek az egyenes rúd végén
display(Math(rf'N_2(y=R) = {latex(N_2)} \qquad V_2(y=R) = {latex(V_2)} \qquad M_{{h2}}(y=R) = {latex(M_h2.subs(y, R))}'))

# Derivált kapcsolatok
display(Math(rf"\frac{{\mathrm{{d}} M_{{h1}}(\varphi)}}{{\mathrm{{d}}(R \varphi)}} = \frac{{1}}{{R}} M_{{h1}}'(\varphi) = {latex(sp.diff(M_h1, phi)/R)} \qquad -V_1(\varphi) = {latex(-V_1)}"))
display(Math(rf"M_{{h2}}'(y) = {latex(sp.diff(M_h2, y))} \qquad -V_2(y) = {latex(-V_2)}"))
```

Látható, hogy az összefüggések teljesülnek.

##### Szélsőértékek

Ahol $V_1(\varphi) = 0$, ott $M_{h1}(\varphi)$-nek szélsőértéke van (vízszintes érintő). Az egyenes rúdon $V_2(y) = F$ állandó, így ott $M_{h2}(y)$ lineáris.

```python
# I. szakasz: V_1(φ) = 0
phi_sz = sp.solve(V_1, phi)
display(Math(rf'V_1(\varphi) = 0 \quad \longrightarrow \quad \varphi = {latex(phi_sz)}'))

# A hajlítónyomaték ezeken a helyeken
for phi_i in phi_sz:
    display(Math(rf'M_{{h1}}(\varphi={latex(phi_i)}) = {latex(M_h1.subs(phi, phi_i))}'))
```

A $\varphi = \pi/2$ hely a kör legfelső pontja, a $\varphi = 3\pi/2$ hely a sarokpont (a tartomány vége).

##### Ábrázolás

Az ábrázoláshoz számértékeket kell helyettesíteni. Az $F = 1$ és $R = 1$ helyettesítéssel a függvények alakja nem változik, a jellemző értékeket pedig szimbolikusan jelöljük.

* A rúd mentén: a függvényértékeket a rúd tengelyvonalára merőlegesen mérjük fel, a görbe rúdon a pozitív értékeket a görbületi középponttól kifelé, az egyenes rúdon $-x$ irányba (mint a kézi rajzon).

* Az ívhossz szerint: a vízszintes tengelyen az ívhossz ($s$) van: a görbe rúdon $s = R \varphi$, az egyenes rúdon $s = \frac{3}{2} R \pi + y$.

```python
def igenybeveteli_abra(ax, f, valtozo, t_0, t_1, s, felirat, lepes=0.1, szin='red'):
    # A függvény kiértékelése numerikusan a [t_0, t_1] tartományon
    t_vals = np.linspace(t_0, t_1, 201)
    f_vals = sp.lambdify(valtozo, f, 'numpy')(t_vals) * np.ones_like(t_vals) # Konstans függvény esetén is tömböt kapunk
    # A vízszintes tengelyen az ívhossz (s) szerint ábrázolunk
    s_vals = sp.lambdify(valtozo, s, 'numpy')(t_vals) * np.ones_like(t_vals)
    # A függvény görbéje (zorder: az s tengely fölé rajzoljuk)
    ax.plot(s_vals, f_vals, color=szin, lw=1.5, zorder=3)
    # Satírozás a görbe és az s tengely között, egyenletes (lepes) távolságonként (mint a kézi rajzokon)
    s_satir = np.arange(s_vals[0], s_vals[-1], lepes)
    ax.vlines(s_satir, 0, np.interp(s_satir, s_vals, f_vals), color=szin, lw=0.8)
    # Lezárás a tartomány két végén
    ax.vlines([s_vals[0], s_vals[-1]], 0, [f_vals[0], f_vals[-1]], color=szin, lw=1.5, zorder=3)
    # Az s tengely
    ax.axhline(0, color='black', lw=1)
    ax.set_ylabel(felirat)
    ax.grid(True, alpha=0.5)

def rud_menti_abra(ax, f, valtozo, t_0, t_1, r, n, skala, lepes=0.1, szin='red'):
    # A függvény, a rúd tengelyvonala (r) és a pozitív oldal felé mutató egységvektor (n) kiértékelése numerikusan
    t_vals = np.linspace(t_0, t_1, 201)
    f_vals = sp.lambdify(valtozo, f, 'numpy')(t_vals) * np.ones_like(t_vals)
    r_vals = np.array([sp.lambdify(valtozo, r_i, 'numpy')(t_vals) * np.ones_like(t_vals) for r_i in r])
    n_vals = np.array([sp.lambdify(valtozo, n_i, 'numpy')(t_vals) * np.ones_like(t_vals) for n_i in n])
    # Az ábra pontjai: a függvényértéket a rúd tengelyvonalára merőlegesen mérjük fel
    p_vals = r_vals + skala*f_vals*n_vals
    # A rúd tengelyvonala
    ax.plot(r_vals[0], r_vals[1], color='black', lw=2)
    # A függvény görbéje (zorder: a rúd fölé rajzoljuk)
    ax.plot(p_vals[0], p_vals[1], color=szin, lw=1.5, zorder=3)
    # Satírozás a görbe és a rúd között, egyenletes (lepes) ívhosszonként (mint a kézi rajzokon)
    ivhossz = np.concatenate([[0], np.cumsum(np.hypot(np.diff(r_vals[0]), np.diff(r_vals[1])))])
    for i in np.searchsorted(ivhossz, np.arange(0, ivhossz[-1], lepes)):
        ax.plot([r_vals[0][i], p_vals[0][i]], [r_vals[1][i], p_vals[1][i]], color=szin, lw=0.8)
    # Lezárás a tartomány két végén
    for i in [0, -1]:
        ax.plot([r_vals[0][i], p_vals[0][i]], [r_vals[1][i], p_vals[1][i]], color=szin, lw=1.5, zorder=3)

def ertek_felirat(ax, szoveg, f, valtozo, t, r, n, skala, eltolas=None, szin='red'):
    # Az ábra pontja a t helyen (kerekítés: a numerikus hiba miatt a 0 ne váltson előjelet)
    f_t = round(float(f.subs(valtozo, t)), 10)
    r_t = np.array([float(r_i.subs(valtozo, t)) for r_i in r])
    n_t = np.array([float(n_i.subs(valtozo, t)) for n_i in n])
    p_t = r_t + skala*f_t*n_t
    ax.plot(p_t[0], p_t[1], 'o', color=szin, ms=3, zorder=10)
    # A feliratot a rúdtól elfelé toljuk el (negatív értéknél a másik oldalra), ha nincs megadva az eltolás
    if eltolas is None:
        eltolas = 5*(n_t if f_t >= 0 else -n_t)
    # Igazítás az eltolás iránya szerint, hogy a felirat ne takarja az ábrát
    ha = 'left' if eltolas[0] > 1.5 else ('right' if eltolas[0] < -1.5 else 'center')
    va = 'bottom' if eltolas[1] > 1.5 else ('top' if eltolas[1] < -1.5 else 'center')
    ax.annotate(szoveg, xy=p_t, xytext=eltolas, textcoords='offset points', ha=ha, va=va, color=szin, fontsize=11)
```

##### Igénybevételi ábrák a rúd mentén

```python
# Számértékek helyettesítése az ábrázoláshoz
ertekek = {F: 1, R: 1}

# A görbe rúd tengelyvonala és a pozitív oldal felé mutató egységvektor (a görbületi középponttól kifelé)
r_1 = sp.Matrix([-R*sp.cos(phi), R*sp.sin(phi)]).subs(ertekek)
n_1 = sp.Matrix([-sp.cos(phi), sp.sin(phi)])

# Az egyenes rúd tengelyvonala és a pozitív oldal felé mutató egységvektor (-x irány)
r_2 = sp.Matrix([0, -R + y]).subs(ertekek)
n_2 = sp.Matrix([-1, 0])

# Az ábrák: függvények a két szakaszon, lépték, cím és a jellemző értékek (szakasz, hely, felirat, a felirat eltolása)
abrak = [((N_1, N_2), 0.3, r'$N(\varphi), ~ N(y)$',
          [(1, 0, r'$0$', None), (1, sp.pi/2, r'$F$', (-4, 4)), (1, sp.pi, r'$0$', None), (1, 3*sp.pi/2, r'$-F$', (4, 5)),
           (2, sp.Rational(1, 2), r'$0$', None)]),
         ((V_1, V_2), 0.3, r'$V(\varphi), ~ V(y)$',
          [(1, 0, r'$F$', None), (1, sp.pi/2, r'$0$', (4, 4)), (1, sp.pi, r'$-F$', (-3, 4)), (1, 3*sp.pi/2, r'$0$', (4, -4)),
           (2, 0, r'$F$', None), (2, 1, r'$F$', (-4, -5))]),
         ((M_h1, M_h2), 0.3, r'$M_h(\varphi), ~ M_h(y)$',
          [(1, 0, r'$0$', None), (1, sp.pi/2, r'$-FR$', (4, -8)), (1, sp.pi, r'$0$', None), (1, 3*sp.pi/2, r'$FR$', None),
           (2, 0, r'$FR$', None), (2, 1, r'$0$', (-4, 4))])]

fig, axs = plt.subplots(1, 3, figsize=(18/2.54, 7.5/2.54))
for ax, ((f_1, f_2), skala, cim, jellemzok) in zip(axs, abrak):
    # Az x és y tengely
    ax.annotate('', xy=(1.75, 0), xytext=(-1.75, 0), arrowprops=dict(arrowstyle='->', color='blue', lw=0.8))
    ax.annotate('', xy=(0, 1.75), xytext=(0, 0), arrowprops=dict(arrowstyle='->', color='blue', lw=0.8))
    ax.annotate(r'$x$', xy=(1.75, 0), xytext=(0, -12), textcoords='offset points', color='blue', fontsize=11)
    ax.annotate(r'$y$', xy=(0, 1.75), xytext=(4, -6), textcoords='offset points', color='blue', fontsize=11)
    # A két szakasz igénybevételi ábrája
    rud_menti_abra(ax, f_1.subs(ertekek), phi, 0, 3*np.pi/2, r_1, n_1, skala, lepes=0.2)
    rud_menti_abra(ax, f_2.subs(ertekek), y, 0, 1, r_2, n_2, skala, lepes=0.2)
    # A jellemző értékek
    for szakasz, t, szoveg, eltolas in jellemzok:
        if szakasz == 1:
            ertek_felirat(ax, szoveg, f_1.subs(ertekek), phi, t, r_1, n_1, skala, eltolas)
        else:
            ertek_felirat(ax, szoveg, f_2.subs(ertekek), y, t, r_2, n_2, skala, eltolas)
    # A φ leíró koordináta iránya: a -x tengelytől, az óramutató járásával megegyezően
    szogek = np.linspace(0.05, 0.8, 30)
    ax.plot(-0.55*np.cos(szogek), 0.55*np.sin(szogek), color='green', lw=1)
    ax.annotate('', xy=(-0.55*np.cos(0.85), 0.55*np.sin(0.85)), xytext=(-0.55*np.cos(0.75), 0.55*np.sin(0.75)),
                arrowprops=dict(arrowstyle='->', color='green'))
    ax.annotate(r'$\varphi$', xy=(-0.4*np.cos(0.4), 0.4*np.sin(0.4)), ha='center', va='center', color='green', fontsize=11)
    ax.set_title(cim)
    ax.set_aspect('equal')
    ax.set_xlim(-1.8, 1.8)
    ax.set_ylim(-1.6, 1.8)
    ax.axis('off')
plt.tight_layout()
plt.savefig('igenybeveteli_abrak_rud_menten.pdf') # Az ábra mentése PDF dokumentumba
plt.show()
```

##### Igénybevételi ábrák az ívhossz szerint

```python
# Az ívhossz a két szakaszon: a görbe rúdon s = R φ, az egyenes rúdon s = 3/2 R π + y
s_1 = (R*phi).subs(ertekek)
s_2 = (sp.Rational(3, 2)*R*sp.pi + y).subs(ertekek)

# A szakaszhatár (sarokpont) és a rúd vége az ívhossz mentén
s_B = float(s_2.subs(y, 0))
s_C = float(s_2.subs(y, 1))

# Az ábrák: függvények a két szakaszon, a tengely felirata és a jellemző értékek (szakasz, hely, felirat, igazítás)
abrak_s = [((N_1, N_2), r'$N$',
            [(1, 0, r'$0$', 'center'), (1, sp.pi/2, r'$F$', 'center'), (1, sp.pi, r'$0$', 'center'), (1, 3*sp.pi/2, r'$-F$', 'center'),
             (2, 0, r'$0$', 'center'), (2, 1, r'$0$', 'center')]),
           ((V_1, V_2), r'$V$',
            [(1, 0, r'$F$', 'center'), (1, sp.pi/2, r'$0$', 'center'), (1, sp.pi, r'$-F$', 'center'), (1, 3*sp.pi/2, r'$0$', 'right'),
             (2, 0, r'$F$', 'left'), (2, 1, r'$F$', 'center')]),
           ((M_h1, M_h2), r'$M_h$',
            [(1, 0, r'$0$', 'center'), (1, sp.pi/2, r'$-FR$', 'center'), (1, sp.pi, r'$0$', 'center'), (1, 3*sp.pi/2, r'$FR$', 'center'),
             (2, 1, r'$0$', 'center')])]

fig, axs = plt.subplots(3, 1, figsize=(16/2.54, 18/2.54), sharex=True)

# A szakaszok jelölése (mint a kézi rajzon)
axs[0].text(s_B/2, 1.05, r'$\mathrm{I.}$', transform=axs[0].get_xaxis_transform(), ha='center', color='green', fontsize=12)
axs[0].text((s_B + s_C)/2, 1.05, r'$\mathrm{II.}$', transform=axs[0].get_xaxis_transform(), ha='center', color='green', fontsize=12)

for ax, ((f_1, f_2), felirat, jellemzok) in zip(axs, abrak_s):
    igenybeveteli_abra(ax, f_1.subs(ertekek), phi, 0, 3*np.pi/2, s_1, felirat, lepes=0.25)
    igenybeveteli_abra(ax, f_2.subs(ertekek), y, 0, 1, s_2, felirat, lepes=0.25)
    # A jellemző értékek (a szakaszhatáron az igazítással kerüljük el, hogy a felirat a függőleges vonalra essen)
    for szakasz, t, szoveg, ha in jellemzok:
        f, valtozo, s = (f_1, phi, s_1) if szakasz == 1 else (f_2, y, s_2)
        s_t = float(s.subs(valtozo, t))
        f_t = round(float(f.subs(ertekek).subs(valtozo, t)), 10)
        dx = {'right': -6, 'center': 0, 'left': 6}[ha]
        ax.plot(s_t, f_t, 'o', color='red', ms=3, zorder=10)
        ax.annotate(szoveg, xy=(s_t, f_t), xytext=(dx, 8 if f_t >= 0 else -14), textcoords='offset points', ha=ha, color='red', fontsize=11)
    ax.set_yticks([0], [r'$0$'])
    ax.set_ylim(-1.6, 1.6)
    # Segédvonalak: a görbe rúd negyedei és a szakaszhatár
    for s_h in [np.pi/2, np.pi, s_B]:
        ax.axvline(s_h, color='gray', ls='--', lw=0.8)

axs[2].set_xticks([0, np.pi/2, np.pi, s_B, s_C], [r'$0$', r'$R \pi / 2$', r'$R \pi$', r'$3 R \pi / 2$', r'$3 R \pi / 2 + R$'])
axs[2].set_xlim(-0.3, s_C + 0.5)
axs[2].set_xlabel(r'$\acute{\imath}\mathrm{vhossz}$') # Az "í" betű nincs benne a cmr10 betűtípusban, ezért mathtext-tel írjuk
plt.tight_layout()
plt.savefig('igenybeveteli_fuggvenyek.pdf') # Az ábra mentése PDF dokumentumba
plt.show()
```
