##### Feladat (Példatár 3.2)

Egy merev testre az $F_1$, $F_2$ és $F_3$ koncentrált erők és az $M_1$ és $M_2$ koncentrált erőpárok hatnak az ábrán feltüntetett módon. Az erők és nyomatékok nagyságai ismertek: $F_1 = 25$ N, $F_2 = 70$ N, $F_3 = 60$ N, $M_1 = 90$ Nm, $M_2 = 110$ Nm. Geometria adatok: $a = 2$ m, $b = 3$ m, $c = 4$ m.

Feladatok:

a) Számítsuk ki az erőrendszer redukáltját az $O$ pontra!

b) Számítsuk ki az erőrendszer redukáltját az $A$ pontra!

c) Igazoljuk, hogy $\mathbf{F} \cdot \mathbf{M}_A$ = $\mathbf{F} \cdot \mathbf{M}_O$!

d) Írjuk fel a centrális egyenes egyenletét és redukáljuk az erőrendszert a centrális egyenes egy tetszőleges pontjára!

e) Milyen $O$ pontban működő erővel és erőpárral kell kiegészíteni az erőrendszert, hogy az egyensúlyi legyen?

##### Csomagok importálása

```python
import numpy as np # Numerikus számítások
import matplotlib.pyplot as plt # Ábrázolás
plt.rcParams['mathtext.fontset'] = 'cm' # Betűtípus az ábrákon
plt.rcParams['font.family'] = 'serif' # Betűtípus az ábrákon
plt.rcParams['font.serif'] = ['cmr10'] # Betűtípus az ábrákon
plt.rcParams['axes.formatter.use_mathtext'] = True # Betűtípus az ábrákon
plt.rcParams['axes.unicode_minus'] = False # Betűtípus az ábrákon
from IPython.display import display, Math, Latex # Szép kimenet Jupyter notebookban
import sympy as sp # Szimbólikus számítások
from sympy import latex # Szép kimenet Jupyter notebookban
sp.init_printing() # Szép kimenet Jupyter notebookban
```

##### Adatok

```python
F_1 = 25 # N
F_2 = 70 # N
F_3 = 60 # N

M_1 = 90 # Nm
M_2 = 110 # Nm

a = 2 # m 
b = 3 # m 
c = 4 # m 
```

##### Az erő és nyomatékvektorok

```python
# Erővektorok
F1_vec = sp.Matrix([0, 0, -F_1])
display(Math(rf'\mathbf{{F}}_1 = {latex(F1_vec)} ~ \mathrm{{N}}'))

F2_vec = sp.Matrix([0, F_2, 0])
display(Math(rf'\mathbf{{F}}_2 = {latex(F2_vec)} ~ \mathrm{{N}}'))

F3_vec = sp.Matrix([-F_3, 0, 0])
display(Math(rf'\mathbf{{F}}_3 = {latex(F3_vec)} ~ \mathrm{{N}}'))

# Nyomatékvektorok
M1_vec = sp.Matrix([M_1, 0, 0])
display(Math(rf'\mathbf{{M}}_1 = {latex(M1_vec)} ~ \mathrm{{Nm}}'))

M2_vec = sp.Matrix([0, M_2, 0])
display(Math(rf'\mathbf{{M}}_2 = {latex(M2_vec)} ~ \mathrm{{Nm}}'))
```

##### Redukálás az $O$ pontba

```python
F = F1_vec + F2_vec + F3_vec
display(Math(rf'\mathbf{{F}} = {latex(F)} ~ \mathrm{{N}}'))
```

```python
# OA helyvektor
r_OA = sp.Matrix([a, b, c])

# OB helyvektor
r_OB = sp.Matrix([0, 0, c])

# OC helyvektor
r_OC = sp.Matrix([a, 0, 0])

M_O = M1_vec + M2_vec + r_OA.cross(F1_vec) + r_OB.cross(F2_vec) + r_OC.cross(F3_vec)
display(Math(rf'\mathbf{{M}}_O = {latex(M_O)} ~ \mathrm{{Nm}}'))
```

##### Redukálás az $A$ pontba

```python
# AO helyvektor
r_AO = -r_OA

M_A = M_O + r_AO.cross(F)
display(Math(rf'\mathbf{{M}}_A = {latex(M_A)} ~ \mathrm{{Nm}}'))
```

##### Invariáns ellenőrzése

```python
eq1 = F.dot(M_O)
eq2 = F.dot(M_A)

display(Math(r'\mathbf{F} \cdot \mathbf{M}_O = \mathbf{F} \cdot \mathbf{M}_A'))
display(Math(rf'{eq1.evalf(5)} ~ \mathrm{{N^2m}}= {eq2.evalf(5)} ~ \mathrm{{N^2m}}'))
```

##### Centrális egyenes

Az origóhoz legközelebbi $G$ pont helyvektora

```python
r_G = F.cross(M_O) / F.norm()**2
display(Math(rf'\mathbf{{r}}_G = {latex(r_G.evalf(5))} ~ \mathrm{{m}}'))
```

A centrális egyenes egységvektora

```python
e_F = F / F.norm()
display(Math(rf'\mathbf{{e}}_F = {latex(e_F.evalf(5))}'))
```

A centrális egyenes egyenlete

```python
lam = sp.Symbol('λ')

r = r_G + lam * e_F
display(Math(rf'\mathbf{{r}} = {latex(r.evalf(5))}'))
```

Redukálás a $G$ pontba

```python
# GO helyevektor
r_GO = - r_G

M_C = M_O + r_GO.cross(F)
display(Math(rf'\mathbf{{M}}_C = {latex(M_C.evalf(5))} ~ \mathrm{{Nm}}'))
```
