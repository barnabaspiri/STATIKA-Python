##### Feladat (Példatár 2.8)

Egy merev testre ható külső erőhatások nyomatéka az $O$ pontra $\mathbf{M}_O = 21\mathbf{i} − 7\mathbf{j} + 14\mathbf{k}$ Nm. Jelöljön ki az $\mathbf{t} = 2\mathbf{i}+3\mathbf{j}+6\mathbf{k}$ vektor egy irányított tengelyt az $O$ ponton keresztül. Mekkora az erre a tengelyre vonatkozó nyomaték?

##### Csomagok importálása

```python
import numpy as np # Numerikus számítások
import matplotlib.pyplot as plt # Ábrázolás
plt.rcParams['mathtext.fontset'] = 'cm' # Betűtípus az ábrákon
plt.rcParams['font.family'] = 'serif' # Betűtípus az ábrákon
plt.rcParams['font.serif'] = ['cmr10'] # Betűtípus az ábrákon
plt.rcParams['axes.formatter.use_mathtext'] = True # Betűtípus az ábrákon
plt.rcParams['axes.unicode_minus'] = False # Betűtípus az ábrákon
from IPython.display import display, Math, Latex # Szép kimenet Jupyter notebookban
import sympy as sp # Szimbólikus számítások
from sympy import latex # Szép kimenet Jupyter notebookban
sp.init_printing() # Szép kimenet Jupyter notebookban
```

##### Adatok

```python
M_x = 21 # Nm 
M_y = -7 # Nm 
M_z = 14 # Nm

t_x = 2 # (1)
t_y = 3 # (1)
t_z = 6 # (1)
```

##### A nyomatékvektor és a kijelölt tengely

```python
M_O = sp.Matrix([M_x, M_y, M_z])
display(Math(rf'\mathbf{{M}}_{{O}}= {latex(M_O)} ~ Nm'))

t = sp.Matrix([t_x, t_y, t_z])
display(Math(rf'\mathbf{{t}}= {latex(t)}'))
```

##### $\mathbf{M}_O$ vetítése a $\mathbf{t}$ tengelyre

```python
# t tengely irányú normálvektor
e_t = t.normalized()
display(Math(rf'\mathbf{{e}}_t = \frac{{\mathbf{{t}}}}{{\vert \mathbf{{t}} \vert }}= {latex(e_t)}'))

# skaláris szorzat
M_t = M_O.dot(e_t)
display(Math(rf'M_t = \mathbf{{M}}_O \cdot \mathbf{{e}}_t = {latex(M_t.evalf(6))} ~ \mathrm{{Nm}}'))
```
